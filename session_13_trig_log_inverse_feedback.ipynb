### Elaboration: absolute values and domain restrictions in trig substitution

---

#### The root cause

The fundamental rule is:

$$\sqrt{u^2} = |u| \qquad \textbf{always}$$

Not $u$. The square root function returns a non-negative result by definition, so $\sqrt{u^2}$ cannot be negative regardless of the sign of $u$. For example:

$$\sqrt{(-3)^2} = \sqrt{9} = 3 = |-3|$$

The step $|u| = u$ is only valid when $u \geq 0$. So whenever you simplify $\sqrt{(\text{trig function})^2}$ in a trig substitution, you must check whether the trig function is non-negative on the domain you have restricted $\theta$ to.

---

#### All three trig substitution patterns

Each pattern comes with a standard domain restriction chosen precisely so that the absolute value drops cleanly.

**Pattern 1 — $\sqrt{a^2 - x^2}$, substitution $x = a\sin\theta$, domain $\theta \in [-\pi/2,\, \pi/2]$**

$$\sqrt{a^2 - x^2} = \sqrt{a^2 - a^2\sin^2\theta} = a\sqrt{1 - \sin^2\theta} = a\sqrt{\cos^2\theta} = a|\cos\theta|$$

On $[-\pi/2, \pi/2]$, cosine is non-negative, so $|\cos\theta| = \cos\theta$:

$$= a\cos\theta$$

**Pattern 2 — $\sqrt{a^2 + x^2}$, substitution $x = a\tan\theta$, domain $\theta \in (-\pi/2,\, \pi/2)$**

$$\sqrt{a^2 + x^2} = a\sqrt{1 + \tan^2\theta} = a\sqrt{\sec^2\theta} = a|\sec\theta|$$

On $(-\pi/2, \pi/2)$, $\cos\theta > 0$, so $\sec\theta > 0$ and $|\sec\theta| = \sec\theta$:

$$= a\sec\theta$$

**Pattern 3 — $\sqrt{x^2 - a^2}$, substitution $x = a\sec\theta$, domain $\theta \in [0,\, \pi/2)$**

$$\sqrt{x^2 - a^2} = a\sqrt{\sec^2\theta - 1} = a\sqrt{\tan^2\theta} = a|\tan\theta|$$

On $[0, \pi/2)$, $\tan\theta \geq 0$, so $|\tan\theta| = \tan\theta$:

$$= a\tan\theta$$

In each case the domain restriction is chosen exactly to make the relevant trig function non-negative. This is not a coincidence — the domain is *defined* by that requirement.

---

#### The other branch of the secant: when does it arise?

The substitution $x = a\sec\theta$ requires $|x| \geq a$, so $x \geq a$ or $x \leq -a$.

- **$x \geq a > 0$:** then $\sec\theta = x/a \geq 1$, which means $\theta \in [0, \pi/2)$. Here $\tan\theta \geq 0$.
- **$x \leq -a < 0$:** then $\sec\theta = x/a \leq -1$, which means $\theta \in (\pi/2, \pi]$. Here $\tan\theta \leq 0$.

If you are evaluating a **definite integral over $x < -a$**, you must use the second branch, and the sign flip matters:

$$\sqrt{\tan^2\theta} = |\tan\theta| = -\tan\theta \qquad \text{(since } \tan\theta \leq 0 \text{ on } (\pi/2, \pi])$$

**Concrete example.** Evaluate $\displaystyle\int_{-5}^{-4} \frac{dx}{\sqrt{x^2-9}}$.

All $x$ values satisfy $x < -3$, so we use the second branch: $x = 3\sec\theta$, $\theta \in (\pi/2, \pi]$.

The substitution gives $dx = 3\sec\theta\tan\theta\,d\theta$ as before. But now:

$$\sqrt{x^2 - 9} = 3|\tan\theta| = -3\tan\theta$$

So the integrand becomes:

$$\frac{3\sec\theta\tan\theta}{-3\tan\theta}\,d\theta = -\sec\theta\,d\theta$$

Integrating: $-\ln|\sec\theta + \tan\theta| + C$.

Back-substituting with the reference triangle (hyp $= |x| = -x$, adj $= 3$, opp $= \sqrt{x^2-9}$; note $x < 0$ so $\sec\theta = x/3 < 0$, $\tan\theta = \sqrt{x^2-9}/x < 0$):

$$-\ln\left|\frac{x}{3} + \frac{\sqrt{x^2-9}}{3}\right| + C = -\ln\left|\frac{x + \sqrt{x^2-9}}{3}\right| + C = -\ln|x + \sqrt{x^2-9}| + \ln 3 + C$$

Absorbing $\ln 3$ into $C$: $-\ln|x + \sqrt{x^2-9}| + C$.

Notice the sign flip relative to the $x > 3$ case — this is consistent, because for $x < -3$ the quantity $x + \sqrt{x^2-9}$ is negative (e.g. at $x=-5$: $-5+4=-1$), so $\ln|x+\sqrt{x^2-9}| = \ln(-(x+\sqrt{x^2-9}))$ is a decreasing function of $x$ while the integral should be positive for $x < -3$.

In practice, **definite integrals over $x < -a$ are rare in a standard calculus course**. When they do appear, the safest approach is to substitute $x = -t$ first to map the integral onto $t > a$, then apply the standard first-branch substitution.

---

#### Practical rules for indefinite integrals

For indefinite integrals (which is the typical exam case), there is a clean protocol:

1. **Always use the first-branch domain** ($\theta \in [0,\pi/2)$ for secant; $\theta \in [-\pi/2,\pi/2]$ for sine; $\theta \in (-\pi/2,\pi/2)$ for tangent). This guarantees the relevant trig function is non-negative and the absolute value drops.

2. **Keep absolute values in the final answer.** Write $\ln|\,\cdot\,|$ rather than $\ln(\,\cdot\,)$ when the argument might be negative (as it is for $x + \sqrt{x^2-9}$ when $x < -3$). The absolute value in the final answer automatically covers both branches.

3. **Never drop an absolute value from $\sqrt{f(\theta)^2}$ unless you have explicitly stated the domain restriction and checked the sign.** When in doubt, write $|\cdot|$ and state your reason for dropping it.

### Elaboration: why the hypotenuse is $\sqrt{x^2+1}$, not $x^2+1$

---

#### What Pythagoras actually says

Pythagoras's theorem relates the **squares** of the sides, not the sides themselves:

$$\text{hypotenuse}^2 = \text{opposite}^2 + \text{adjacent}^2$$

If opposite $= x$ and adjacent $= 1$, then:

$$\text{hypotenuse}^2 = x^2 + 1^2 = x^2 + 1$$

So $x^2 + 1$ is the **square** of the hypotenuse. To get the hypotenuse itself, take the square root:

$$\text{hypotenuse} = \sqrt{x^2 + 1}$$

Labelling the hypotenuse as $x^2+1$ is the same mistake as saying "the side of a square with area 9 is 9" — the area formula gives you the **square** of the side, so the side is $\sqrt{9} = 3$.

---

#### A concrete numerical check

Take $x = 3$. With opposite $= 3$, adjacent $= 1$:

- **Your labelling:** hypotenuse $= x^2+1 = 10$. Triangle has sides $3$, $1$, $10$.
- **Pythagoras check:** $3^2 + 1^2 = 9 + 1 = 10$. But this equals $10$, not $10^2 = 100$. So $3^2 + 1^2 \neq 10^2$ — a triangle with sides $3, 1, 10$ is **not** a right triangle. In fact $3 + 1 = 4 < 10$, so these three lengths cannot even form a triangle at all.
- **Correct labelling:** hypotenuse $= \sqrt{x^2+1} = \sqrt{10} \approx 3.16$. Check: $3^2 + 1^2 = 10 = (\sqrt{10})^2$ ✓

The numerical check is a fast sanity test you can always use: plug in a specific value of $x$ and verify $\text{opp}^2 + \text{adj}^2 = \text{hyp}^2$.

---

#### The consequence for $\sin\theta$ and $\cos\theta$

With the correct hypotenuse $\sqrt{x^2+1}$:

$$\sin\theta = \frac{x}{\sqrt{x^2+1}}, \qquad \cos\theta = \frac{1}{\sqrt{x^2+1}}$$

Quick identity check — these must satisfy $\sin^2\theta + \cos^2\theta = 1$:

$$\frac{x^2}{x^2+1} + \frac{1}{x^2+1} = \frac{x^2+1}{x^2+1} = 1 \quad \checkmark$$

With the wrong hypotenuse $x^2+1$:

$$\frac{x^2}{(x^2+1)^2} + \frac{1}{(x^2+1)^2} = \frac{x^2+1}{(x^2+1)^2} = \frac{1}{x^2+1} \neq 1 \quad \text{(for } x \neq 0\text{)}$$

If the values you read off the triangle do not satisfy $\sin^2\theta + \cos^2\theta = 1$, the triangle is labelled incorrectly. This is a reliable self-check.

---

#### The general pattern for all three substitutions

| Substitution | Triangle sides | Hypotenuse |
|---|---|---|
| $x = a\sin\theta$ | opp $= x$, hyp $= a$ | adj $= \sqrt{a^2 - x^2}$ (solve for missing side) |
| $x = a\tan\theta$ | opp $= x$, adj $= a$ | hyp $= \sqrt{x^2 + a^2}$ |
| $x = a\sec\theta$ | hyp $= x$, adj $= a$ | opp $= \sqrt{x^2 - a^2}$ |

In every case: Pythagoras gives you the **square** of the missing side. Take the square root to get the side length itself, then read off the trig ratios.
---

#### Common confusion: "the denominator is $(x^2+1)^2$, so isn't $\sqrt{(x^2+1)^2} = x^2+1$?"

Yes — $\sqrt{(x^2+1)^2} = x^2+1$ is correct arithmetic. But this calculation is irrelevant to the reference triangle, because $(x^2+1)^2$ never appears under a square root in the triangle construction.

The reference triangle is built entirely from the **substitution** $x = \tan\theta$:

$$\text{opposite} = x, \quad \text{adjacent} = 1$$

Pythagoras squares **the two legs** and sums them:

$$\text{hypotenuse} = \sqrt{\underbrace{x^2}_{\text{opp}^2} + \underbrace{1^2}_{\text{adj}^2}} = \sqrt{x^2 + 1}$$

What is under the square root is $x^2 + 1$ — the sum of two squares — not $(x^2+1)^2$.

The $(x^2+1)^2$ in the denominator of the original integral is a separate object. It was already dealt with during the substitution step, where it became $\sec^4\theta$. By the time you reach back-substitution, it no longer appears anywhere in the triangle calculation.

**The triangle's side lengths depend only on which substitution you used, never on the structure of the integrand.**
---

#### How to construct the reference triangle: the right heuristic

The substitution itself tells you which two sides to label. Pythagoras gives you the third. The integrand plays no role in the triangle — it only told you which substitution to reach for in the first place.

**Step 1 — Read two sides off the substitution.**

Each substitution is a trig ratio, and a trig ratio is a fraction of two sides:

| Substitution | Trig ratio | Two sides labelled directly |
|---|---|---|
| $x = a\sin\theta$ | $\sin\theta = \dfrac{x}{a} = \dfrac{\text{opp}}{\text{hyp}}$ | opposite $= x$, hypotenuse $= a$ |
| $x = a\tan\theta$ | $\tan\theta = \dfrac{x}{a} = \dfrac{\text{opp}}{\text{adj}}$ | opposite $= x$, adjacent $= a$ |
| $x = a\sec\theta$ | $\sec\theta = \dfrac{x}{a} = \dfrac{\text{hyp}}{\text{adj}}$ | hypotenuse $= x$, adjacent $= a$ |

**Step 2 — Use Pythagoras to find the remaining side.**

Square the two known sides, apply $\text{hyp}^2 = \text{opp}^2 + \text{adj}^2$, and take the square root:

| Substitution | Known sides | Remaining side |
|---|---|---|
| $x = a\sin\theta$ | opp $= x$, hyp $= a$ | adj $= \sqrt{a^2 - x^2}$ |
| $x = a\tan\theta$ | opp $= x$, adj $= a$ | hyp $= \sqrt{x^2 + a^2}$ |
| $x = a\sec\theta$ | hyp $= x$, adj $= a$ | opp $= \sqrt{x^2 - a^2}$ |

**Step 3 — Read off any trig ratio you need from the completed triangle.**

For $x = \tan\theta$ (the Q3 case, $a = 1$): opp $= x$, adj $= 1$, hyp $= \sqrt{x^2+1}$. Every back-substitution value follows immediately:

$$\sin\theta = \frac{x}{\sqrt{x^2+1}}, \quad \cos\theta = \frac{1}{\sqrt{x^2+1}}, \quad \tan\theta = x, \quad \sec\theta = \sqrt{x^2+1}$$

The structure of the integrand — whether it was $\frac{1}{x^2+1}$, $\frac{1}{(x^2+1)^2}$, or $\sqrt{x^2+1}$ — is irrelevant. The triangle is the same in every case, because the triangle comes from the substitution, not the integrand.

### Elaboration: recognising trig substitution situations in the wild

The three trig substitution patterns all hinge on the same core idea: get the expression inside the root (or the quadratic factor) into one of the forms $a^2 - u^2$, $a^2 + u^2$, or $u^2 - a^2$. Any manipulation you do beforehand is just preparation to reach one of those three forms.

---

#### Step 1 — Identify the candidate expression

Trig substitution is relevant when the integrand contains:

| Form | Note |
|---|---|
| $\sqrt{\text{quadratic in }x}$ | The classic signal |
| $(\text{quadratic in }x)^{n}$ for integer $n \geq 2$ in the denominator | No root needed — the identity still collapses it (as in Q3) |
| $(\text{quadratic in }x)^{n/2}$ for half-integer $n$ | Same idea |

If the quadratic is linear in $x$ as well (i.e., has both an $x$ and an $x^2$ term), manipulation is needed first.

---

#### Step 2 — Ask: does the quadratic have a linear term?

Look at the expression inside the root or power. Count the types of $x$ terms present:

| What you see | What to do |
|---|---|
| Only $x^2$ (no $x$ term) | Go straight to Step 3 — may already be in standard form |
| Both $x^2$ and $x$ | **Complete the square** to eliminate the linear term |
| Only $x$ (no $x^2$) | Not a trig sub — probably a simple substitution |

**Completing the square — the mechanical procedure:**

Given $\pm x^2 + bx + c$:

1. Factor out the leading coefficient if it isn't $\pm 1$.
2. Write $x^2 + bx = (x + b/2)^2 - (b/2)^2$.
3. The result is $(x + b/2)^2 + \text{constant}$ — the constant's sign tells you which pattern applies.

**Example:** $\sqrt{3 + 2x - x^2}$
$$3 + 2x - x^2 = -(x^2 - 2x) + 3 = -\bigl((x-1)^2 - 1\bigr) + 3 = 4 - (x-1)^2$$
Pattern: $a^2 - u^2$ with $a = 2$, $u = x-1$. Substitution: $x - 1 = 2\sin\theta$.

**Example:** $\sqrt{x^2 + 6x + 13}$
$$x^2 + 6x + 13 = (x+3)^2 + 4$$
Pattern: $u^2 + a^2$ with $u = x+3$, $a = 2$. Substitution: $x + 3 = 2\tan\theta$.

---

#### Step 3 — If no linear term: check for a coefficient mismatch

If the quadratic is $\alpha x^2 + \beta$ (no linear term), it won't match the standard form $a^2 \pm u^2$ unless the coefficient on $x^2$ is $1$. Factor it into standard form:

$$\sqrt{9 - 4x^2} = \sqrt{9\left(1 - \tfrac{4}{9}x^2\right)} = 3\sqrt{1 - \left(\tfrac{2}{3}x\right)^2}$$

Now it's $a^2 - u^2$ with $a = 1$, $u = \frac{2}{3}x$. Substitution: $\frac{2}{3}x = \sin\theta$, i.e. $x = \frac{3}{2}\sin\theta$.

Alternatively, recognise $\sqrt{a^2 - (bx)^2}$ directly and substitute $bx = a\sin\theta$:
$$\sqrt{9 - 4x^2} = \sqrt{3^2 - (2x)^2} \quad \Rightarrow \quad 2x = 3\sin\theta$$
Both approaches give the same substitution.

---

#### Step 4 — Read the sign to choose the pattern

Once you have the form $a^2 \pm u^2$ or $u^2 - a^2$, the sign structure tells you which substitution to use — and there's a geometric way to remember it:

| Form | When is it positive? | Substitution | Why it fits |
|---|---|---|---|
| $a^2 - u^2$ | Only when $\|u\| < a$ — a **bounded** range | $u = a\sin\theta$ | $\sin\theta$ is bounded: $\|\sin\theta\| \leq 1$ |
| $a^2 + u^2$ | **Always** — for all real $u$ | $u = a\tan\theta$ | $\tan\theta$ ranges over all of $\mathbb{R}$ |
| $u^2 - a^2$ | Only when $\|u\| > a$ — an **unbounded** range away from zero | $u = a\sec\theta$ | $\|\sec\theta\| \geq 1$ |

The substitution's range matches the domain on which the expression is positive. This is why the signs and the substitutions go together the way they do.

---

#### Step 5 — No square root? Check if the pattern still applies

If the integrand has $(x^2 + a^2)^n$ or $(a^2 - x^2)^n$ for an integer power $n \geq 2$ — with **no** square root — trig substitution can still be the right tool. The substitution turns the polynomial into a power of a single trig function, which then simplifies via identities.

Signal: the denominator is a power of a sum/difference of squares, and no other method (partial fractions, simple $u$-substitution) collapses it cleanly.

Q3 is the example: $\dfrac{1}{(x^2+1)^2}$ → $x = \tan\theta$ turns the denominator into $\sec^4\theta$, which cancels with $dx = \sec^2\theta\,d\theta$ to leave $\cos^2\theta$.

---

#### Diagnostic checklist

When you see a quadratic expression inside a root or raised to a power, run through these questions in order:

1. **Is there a linear $x$ term?** → Yes: complete the square. No: continue.
2. **Is the coefficient on $x^2$ equal to $1$?** → No: factor it out or absorb it into the substitution.
3. **What is the sign pattern?** → $a^2 - u^2$: sine. $a^2 + u^2$: tangent. $u^2 - a^2$: secant.
4. **Is there a square root?** → No: check if integer-power trig sub would collapse it. Yes: proceed normally.
5. **What is $u$?** → Usually $x$ or $(x - h)$ after completing the square. The substitution is $u = a \cdot \text{trig}(\theta)$.

After step 5, the reference triangle heuristic takes over: the substitution labels two sides, Pythagoras gives the third.

## Q5 Feedback — $\int \dfrac{3x^2 + 2}{x^3 + 2x + 1}\,dx$

**Correct.** You recognised the pattern immediately and wrote down the result in one step — that is exactly the right approach here.

$$\frac{d}{dx}(x^3 + 2x + 1) = 3x^2 + 2 \quad \Rightarrow \quad \int \frac{3x^2+2}{x^3+2x+1}\,dx = \ln|x^3+2x+1| + C \quad \checkmark$$

The core pattern for all of Section B is $\displaystyle\int \frac{f'(x)}{f(x)}\,dx = \ln|f(x)| + C$. Q5 is a clean instance: the numerator is exactly the derivative of the denominator, so no manipulation is needed before writing down the answer.

Q6 and Q7 require a small step of preparation before the same pattern applies — either rewriting the integrand to expose the ratio, or choosing the right substitution variable.

## Q6 Feedback — $\int \tan x\,dx$

**Correct.** The key insight is there: the numerator $\sin x$ is $-\frac{d}{dx}\cos x$, so the integral is $-\ln|\cos x| + C$.

---

### One notation point

The line
$$\int \frac{\sin x}{\cos x} = \frac{-f'(x)}{f(x)}$$
mixes an integral on the left with a plain expression on the right. What you mean is that the integrand *equals* $-f'(x)/f(x)$, so the pattern applies:
$$\frac{\sin x}{\cos x} = \frac{-(-\sin x)}{\cos x} = \frac{-f'(x)}{f(x)} \quad \Rightarrow \quad \int \frac{\sin x}{\cos x}\,dx = -\ln|\cos x| + C$$
A cleaner way to write the step is to introduce the minus sign explicitly before applying the pattern:
$$\int \frac{\sin x}{\cos x}\,dx = -\int \frac{-\sin x}{\cos x}\,dx = -\int \frac{(\cos x)'}{\cos x}\,dx = -\ln|\cos x| + C$$

---

### Alternative form

$-\ln|\cos x| + C$ is correct and complete. It is also commonly written as:
$$-\ln|\cos x| + C = \ln|\cos x|^{-1} + C = \ln|\sec x| + C$$
Both forms appear in textbooks; $\ln|\sec x| + C$ is more standard. Either is acceptable.
