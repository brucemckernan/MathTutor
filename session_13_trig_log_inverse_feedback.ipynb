### Elaboration: absolute values and domain restrictions in trig substitution

---

#### The root cause

The fundamental rule is:

$$\sqrt{u^2} = |u| \qquad \textbf{always}$$

Not $u$. The square root function returns a non-negative result by definition, so $\sqrt{u^2}$ cannot be negative regardless of the sign of $u$. For example:

$$\sqrt{(-3)^2} = \sqrt{9} = 3 = |-3|$$

The step $|u| = u$ is only valid when $u \geq 0$. So whenever you simplify $\sqrt{(\text{trig function})^2}$ in a trig substitution, you must check whether the trig function is non-negative on the domain you have restricted $\theta$ to.

---

#### All three trig substitution patterns

Each pattern comes with a standard domain restriction chosen precisely so that the absolute value drops cleanly.

**Pattern 1 — $\sqrt{a^2 - x^2}$, substitution $x = a\sin\theta$, domain $\theta \in [-\pi/2,\, \pi/2]$**

$$\sqrt{a^2 - x^2} = \sqrt{a^2 - a^2\sin^2\theta} = a\sqrt{1 - \sin^2\theta} = a\sqrt{\cos^2\theta} = a|\cos\theta|$$

On $[-\pi/2, \pi/2]$, cosine is non-negative, so $|\cos\theta| = \cos\theta$:

$$= a\cos\theta$$

**Pattern 2 — $\sqrt{a^2 + x^2}$, substitution $x = a\tan\theta$, domain $\theta \in (-\pi/2,\, \pi/2)$**

$$\sqrt{a^2 + x^2} = a\sqrt{1 + \tan^2\theta} = a\sqrt{\sec^2\theta} = a|\sec\theta|$$

On $(-\pi/2, \pi/2)$, $\cos\theta > 0$, so $\sec\theta > 0$ and $|\sec\theta| = \sec\theta$:

$$= a\sec\theta$$

**Pattern 3 — $\sqrt{x^2 - a^2}$, substitution $x = a\sec\theta$, domain $\theta \in [0,\, \pi/2)$**

$$\sqrt{x^2 - a^2} = a\sqrt{\sec^2\theta - 1} = a\sqrt{\tan^2\theta} = a|\tan\theta|$$

On $[0, \pi/2)$, $\tan\theta \geq 0$, so $|\tan\theta| = \tan\theta$:

$$= a\tan\theta$$

In each case the domain restriction is chosen exactly to make the relevant trig function non-negative. This is not a coincidence — the domain is *defined* by that requirement.

---

#### The other branch of the secant: when does it arise?

The substitution $x = a\sec\theta$ requires $|x| \geq a$, so $x \geq a$ or $x \leq -a$.

- **$x \geq a > 0$:** then $\sec\theta = x/a \geq 1$, which means $\theta \in [0, \pi/2)$. Here $\tan\theta \geq 0$.
- **$x \leq -a < 0$:** then $\sec\theta = x/a \leq -1$, which means $\theta \in (\pi/2, \pi]$. Here $\tan\theta \leq 0$.

If you are evaluating a **definite integral over $x < -a$**, you must use the second branch, and the sign flip matters:

$$\sqrt{\tan^2\theta} = |\tan\theta| = -\tan\theta \qquad \text{(since } \tan\theta \leq 0 \text{ on } (\pi/2, \pi])$$

**Concrete example.** Evaluate $\displaystyle\int_{-5}^{-4} \frac{dx}{\sqrt{x^2-9}}$.

All $x$ values satisfy $x < -3$, so we use the second branch: $x = 3\sec\theta$, $\theta \in (\pi/2, \pi]$.

The substitution gives $dx = 3\sec\theta\tan\theta\,d\theta$ as before. But now:

$$\sqrt{x^2 - 9} = 3|\tan\theta| = -3\tan\theta$$

So the integrand becomes:

$$\frac{3\sec\theta\tan\theta}{-3\tan\theta}\,d\theta = -\sec\theta\,d\theta$$

Integrating: $-\ln|\sec\theta + \tan\theta| + C$.

Back-substituting with the reference triangle (hyp $= |x| = -x$, adj $= 3$, opp $= \sqrt{x^2-9}$; note $x < 0$ so $\sec\theta = x/3 < 0$, $\tan\theta = \sqrt{x^2-9}/x < 0$):

$$-\ln\left|\frac{x}{3} + \frac{\sqrt{x^2-9}}{3}\right| + C = -\ln\left|\frac{x + \sqrt{x^2-9}}{3}\right| + C = -\ln|x + \sqrt{x^2-9}| + \ln 3 + C$$

Absorbing $\ln 3$ into $C$: $-\ln|x + \sqrt{x^2-9}| + C$.

Notice the sign flip relative to the $x > 3$ case — this is consistent, because for $x < -3$ the quantity $x + \sqrt{x^2-9}$ is negative (e.g. at $x=-5$: $-5+4=-1$), so $\ln|x+\sqrt{x^2-9}| = \ln(-(x+\sqrt{x^2-9}))$ is a decreasing function of $x$ while the integral should be positive for $x < -3$.

In practice, **definite integrals over $x < -a$ are rare in a standard calculus course**. When they do appear, the safest approach is to substitute $x = -t$ first to map the integral onto $t > a$, then apply the standard first-branch substitution.

---

#### Practical rules for indefinite integrals

For indefinite integrals (which is the typical exam case), there is a clean protocol:

1. **Always use the first-branch domain** ($\theta \in [0,\pi/2)$ for secant; $\theta \in [-\pi/2,\pi/2]$ for sine; $\theta \in (-\pi/2,\pi/2)$ for tangent). This guarantees the relevant trig function is non-negative and the absolute value drops.

2. **Keep absolute values in the final answer.** Write $\ln|\,\cdot\,|$ rather than $\ln(\,\cdot\,)$ when the argument might be negative (as it is for $x + \sqrt{x^2-9}$ when $x < -3$). The absolute value in the final answer automatically covers both branches.

3. **Never drop an absolute value from $\sqrt{f(\theta)^2}$ unless you have explicitly stated the domain restriction and checked the sign.** When in doubt, write $|\cdot|$ and state your reason for dropping it.
