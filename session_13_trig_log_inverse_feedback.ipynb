## Q1 Feedback — $\int \dfrac{dx}{\sqrt{x^2 - 9}}$

**Correct.** Clean execution all the way through.

### What you did well

- Correctly identified the secant pattern and set $x = 3\sec\theta$.
- Computed $dx = 3\sec\theta\tan\theta\,d\theta$ correctly.
- Factored $9$ out from under the root cleanly: $\sqrt{9(\sec^2\theta - 1)} = 3\sqrt{\sec^2\theta - 1}$.
- Applied $\sec^2\theta - 1 = \tan^2\theta$ and cancelled $\tan\theta$ top and bottom to reach $\int \sec\theta\,d\theta$.
- Quoted $\int \sec\theta\,d\theta = \ln|\sec\theta + \tan\theta| + C$ correctly.
- Back-substituted $\sec\theta = x/3$ and $\tan\theta = \sqrt{x^2-9}/3$ correctly.
- Absorbed $-\ln 3$ into the constant at the end — this is the right move.

---

### One implicit step worth making explicit

At the line where you write $\sqrt{\sec^2\theta - 1} = \tan\theta$, you are using:

$$\sqrt{\tan^2\theta} = |\tan\theta| = \tan\theta$$

The last equality holds because $\theta \in [0, \pi/2)$, where $\tan\theta \geq 0$. This is the reason the domain restriction $\theta \in [0, \pi/2)$ is stated in the question — it ensures the absolute value drops cleanly. If you used $\theta \in (\pi/2, \pi]$ instead (the other branch of $\sec\theta \geq 1$), you would need $|\tan\theta| = -\tan\theta$ and the signs would need care.

In practice: always check the domain restriction is in place before dropping the absolute value from a square root.

---

### The reference triangle

Your back-substitution is correct. The underlying triangle: since $\sec\theta = x/3$ and $\sec\theta = \text{hypotenuse}/\text{adjacent}$, label the triangle with hypotenuse $x$ and adjacent side $3$. By Pythagoras the opposite side is $\sqrt{x^2 - 9}$, giving:

$$\tan\theta = \frac{\text{opp}}{\text{adj}} = \frac{\sqrt{x^2-9}}{3}$$

Drawing this triangle and reading off $\tan\theta$ (and any other trig ratios needed) is the standard back-substitution strategy for all three trig-sub patterns.

---

### Verification

Differentiate the answer to confirm:

$$\frac{d}{dx}\ln\left|x + \sqrt{x^2 - 9}\right| = \frac{1 + \dfrac{x}{\sqrt{x^2-9}}}{x + \sqrt{x^2-9}} = \frac{\dfrac{\sqrt{x^2-9}+x}{\sqrt{x^2-9}}}{x + \sqrt{x^2-9}} = \frac{1}{\sqrt{x^2-9}} \quad \checkmark$$

*(Alternative form: the answer can also be written $\operatorname{arccosh}(x/3) + C$, since $\operatorname{arccosh}(u) = \ln|u + \sqrt{u^2-1}|$. The two forms are identical up to a constant.)*
