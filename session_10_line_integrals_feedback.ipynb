## Q1 — Line integral of a scalar field

### Result: Correct ✓ — clean throughout

Every step is right. Here is a brief confirmation of what each step is doing and why it is necessary, since this is the first scalar line integral of the session.

---

**Parametrisation** $x = t,\ y = 2t,\ t \in [0,1]$

A line segment from $(0,0)$ to $(1,2)$ can always be written as $\mathbf{r}(t) = (1-t)(0,0) + t(1,2) = (t, 2t)$. You chose this directly. ✓

**Arc-length element $ds$**

$$ds = |\mathbf{r}'(t)|\,dt = \sqrt{1^2 + 2^2}\,dt = \sqrt{5}\,dt$$

This is the key step that distinguishes a scalar line integral from a plain integral. $ds$ measures actual arc length along the curve, not displacement in $t$. On a straight line with constant speed $|\mathbf{r}'| = \sqrt{5}$, the arc-length element is just $\sqrt{5}\,dt$. ✓

**Substitution of $x$ and $y$**

$$x + y^2 = t + (2t)^2 = t + 4t^2 \quad \checkmark$$

**Evaluation**

$$\sqrt{5}\int_0^1 (t + 4t^2)\,dt = \sqrt{5}\left[\frac{t^2}{2} + \frac{4t^3}{3}\right]_0^1 = \sqrt{5}\left(\frac{1}{2} + \frac{4}{3}\right) = \sqrt{5}\cdot\frac{11}{6} \quad \checkmark$$

---

**Conceptual note: what this integral computes**

A scalar line integral $\int_C f\,ds$ weights the value of $f$ by arc length as you travel along $C$. Geometrically, if $f(x,y)$ is the height of a fence built above the curve $C$, then $\int_C f\,ds$ is the area of that fence. The $\sqrt{5}$ factor reflects that the curve has length $\sqrt{5}$ (not 1) — every infinitesimal piece of $C$ contributes more area than a flat $dt$ step would suggest.

## Q2 — Conservative fields

### Result: Correct ✓ — all three parts clean, one terminology note

---

### Part (a) — Conservative check ✓

$$\frac{\partial P}{\partial y} = \frac{\partial}{\partial y}(2xy + y^3) = 2x + 3y^2, \qquad \frac{\partial Q}{\partial x} = \frac{\partial}{\partial x}(x^2 + 3xy^2) = 2x + 3y^2$$

Equal, so $\mathbf{F}$ is conservative. ✓

**Terminology note — "zero curl" vs "equal cross-partials"**

In 2D, the correct condition to state is $\partial P/\partial y = \partial Q/\partial x$. Saying "zero curl" is not wrong in spirit — in 3D the curl of $\mathbf{F} = (P, Q, 0)$ is $\left(\frac{\partial Q}{\partial x} - \frac{\partial P}{\partial y}\right)\hat{k}$, so the 2D condition is indeed the z-component of curl being zero. But "zero curl" is imprecise when working in 2D, and in an exam it is safer to state the condition directly as equality of cross-partials.

---

### Part (b) — Potential function ✓

Your method is exactly right:

1. Integrate $P = 2xy + y^3$ with respect to $x$:
$$\varphi(x,y) = x^2y + xy^3 + g(y)$$

2. Differentiate with respect to $y$ and match to $Q$:
$$\frac{\partial \varphi}{\partial y} = x^2 + 3xy^2 + g'(y) = x^2 + 3xy^2 \implies g'(y) = 0 \implies g(y) = C$$

3. Potential function: $\varphi(x,y) = x^2y + xy^3 + C$ ✓

Your explanation of why $g(y) = C$ is also correct and well-reasoned: because the $y$-information was already fully captured in the integration step, nothing is left for $g(y)$ to contribute.

---

### Part (c) — FTLI ✓

$$\int_C \mathbf{F}\cdot d\mathbf{r} = \varphi(1,2) - \varphi(0,0) = (1^2\cdot2 + 1\cdot2^3 + C) - (C) = (2 + 8) - 0 = 10 \quad \checkmark$$

The constant $C$ cancels cleanly, as it always must — the potential function is only defined up to a constant, and the FTLI only ever uses the difference $\varphi(b) - \varphi(a)$.

**Key point you correctly identified:** path independence means any path from $(0,0)$ to $(1,2)$ gives the same answer. You could go along a straight line, a spiral, or any curve you like — the answer is always 10. This is the main payoff of identifying a conservative field before integrating.

## Q3 — Non-conservative field

### Result: Correct answer ✓ — one sign error mid-way, recovered incorrectly

The final answer $2\pi$ is right, but there is a sign error in the dot product step that is then "fixed" with flawed reasoning. Both need to be noted.

---

### Conservative check ✓

$$P = -y,\quad Q = x, \qquad \frac{\partial P}{\partial y} = -1 \neq 1 = \frac{\partial Q}{\partial x}$$

Cross-partials are unequal, so $\mathbf{F}$ is not conservative. ✓

---

### Parametrisation ✓

$$\mathbf{r}(t) = \cos t\,\hat{\imath} + \sin t\,\hat{\jmath}, \quad t \in [0, 2\pi]$$
$$\mathbf{r}'(t) = -\sin t\,\hat{\imath} + \cos t\,\hat{\jmath} \quad \checkmark$$

---

### Substituting into $\mathbf{F}$ ✓

$$\mathbf{F}(x(t), y(t)) = (-\sin t,\; \cos t) \quad \checkmark$$

---

### Dot product — sign error here ✗

You wrote:

$$\mathbf{F} \cdot \mathbf{r}'= (-\sin t)(-\sin t) + (\cos t)(\cos t) = -\sin^2 t + \cos^2 t$$

The **first term is wrong**. $(-\sin t)(-\sin t) = +\sin^2 t$, not $-\sin^2 t$. The two negatives multiply to give a positive. The correct dot product is:

$$\mathbf{F} \cdot \mathbf{r}' = \sin^2 t + \cos^2 t = 1$$

No further manipulation is needed.

---

### The recovery step is also invalid

You wrote: *"I can drop the leading minus on sin as it is squared and therefore always positive."*

This reasoning does not hold. A term being always positive does not let you drop a minus sign — $-\sin^2 t$ is always non-positive, not always non-negative. You cannot convert $-\sin^2 t + \cos^2 t$ to $\sin^2 t + \cos^2 t$ on this basis.

What actually happened: you made a sign error (writing $-\sin^2 t$ instead of $+\sin^2 t$), then made a second sign error in the "fix", and the two errors cancelled, leaving the correct integrand $\sin^2 t + \cos^2 t = 1$.

The correct single-step reasoning is:

$$\mathbf{F} \cdot \mathbf{r}' = (-\sin t)(\underbrace{-\sin t}_{r'_x}) + (\cos t)(\underbrace{\cos t}_{r'_y}) = \sin^2 t + \cos^2 t = 1$$

---

### Evaluation ✓

$$\oint_C \mathbf{F}\cdot d\mathbf{r} = \int_0^{2\pi} 1\,dt = 2\pi \quad \checkmark$$

---

### Conceptual note — why $2\pi$ for this field

$\mathbf{F} = (-y, x)$ is the field that points tangentially around every circle centred at the origin. On the unit circle, $\mathbf{F}$ and $\mathbf{r}'$ are **identical** (both equal $(-\sin t, \cos t)$), so $\mathbf{F}\cdot\mathbf{r}' = |\mathbf{r}'|^2 = 1$ at every point. The work done per unit parameter is constantly 1 over an interval of length $2\pi$, giving exactly $2\pi$.

This also equals the area of the unit circle ($\pi \cdot 1^2 = \pi$) multiplied by 2 — which is no coincidence: by Green's Theorem,

$$\oint_C (-y\,dx + x\,dy) = \iint_D \left(\frac{\partial Q}{\partial x} - \frac{\partial P}{\partial y}\right) dA = \iint_D (1 - (-1))\,dA = 2 \cdot \text{Area}(D) = 2\pi$$

You will use this in Q4.

## Q4 — Green's Theorem

### Result: Method fully correct ✓ — algebra error in the final simplification gives wrong answer

The approach is excellent: Green's Theorem is correctly stated and applied, the region is correctly identified, and the inner integral is correctly evaluated. One term-dropping error in the outer integrand produces the wrong final answer.

---

### Green's Theorem — correctly applied ✓

$$\oint_C P\,dx + Q\,dy = \iint_D \left(\frac{\partial Q}{\partial x} - \frac{\partial P}{\partial y}\right) dA$$

With $P = y^2$, $Q = x$:

$$\frac{\partial Q}{\partial x} = 1, \qquad \frac{\partial P}{\partial y} = 2y, \qquad \frac{\partial Q}{\partial x} - \frac{\partial P}{\partial y} = 1 - 2y \quad \checkmark$$

---

### Region ✓

Intersection of $y = x$ and $y = x^2$: solve $x = x^2 \Rightarrow x(x-1)=0$, so $(0,0)$ and $(1,1)$. ✓

At $x = 0.5$: $y = x = 0.5 > y = x^2 = 0.25$, so $y = x^2$ is the lower boundary. ✓

$$\iint_D = \int_0^1 \int_{x^2}^{x} (1 - 2y)\,dy\,dx \quad \checkmark$$

---

### Inner integral ✓

$$\int_{y=x^2}^{y=x}(1-2y)\,dy = \Big[y - y^2\Big]_{y=x^2}^{y=x} = (x - x^2) - (x^2 - x^4) = x - 2x^2 + x^4 \quad \checkmark$$

---

### Outer integral — algebra error here ✗

You wrote:

$$\int_0^1 \underbrace{x - x^2 - x^2 + x^4}_{\text{correct}} \,dx = \int_0^1 \underbrace{x}_{\text{wrong}}\,dx$$

The $-2x^2 + x^4$ terms were dropped without justification. The correct outer integral is:

$$\int_0^1 (x - 2x^2 + x^4)\,dx = \left[\frac{x^2}{2} - \frac{2x^3}{3} + \frac{x^5}{5}\right]_0^1 = \frac{1}{2} - \frac{2}{3} + \frac{1}{5}$$

Finding a common denominator of 30:

$$= \frac{15}{30} - \frac{20}{30} + \frac{6}{30} = \frac{1}{30}$$

The correct answer is $\boxed{\dfrac{1}{30}}$, not $\dfrac{1}{2}$.

---

### Side note on the conservativeness check

You checked that $\partial P/\partial y \neq \partial Q/\partial x$ at the start. This is sensible intuition — it tells you the answer won't be zero (a conservative field integrates to 0 around any closed curve, so you'd know immediately if something went wrong). But Green's Theorem applies whether or not $\mathbf{F}$ is conservative; the check isn't required to use it.
