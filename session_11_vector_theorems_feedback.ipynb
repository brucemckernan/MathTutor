## Q1 — Curl and Stokes' Theorem

### Part (a) — Curl ✓

With $\mathbf{F} = (-y, x, z^2)$, the curl determinant expands as:

$$\nabla \times \mathbf{F} =
\left(\frac{\partial z^2}{\partial y} - \frac{\partial x}{\partial z}\right)\hat{\imath}
-\left(\frac{\partial z^2}{\partial x} - \frac{\partial(-y)}{\partial z}\right)\hat{\jmath}
+\left(\frac{\partial x}{\partial x} - \frac{\partial(-y)}{\partial y}\right)\hat{k}$$

$$= (0 - 0)\hat{\imath} - (0 - 0)\hat{\jmath} + (1 - (-1))\hat{k} = 2\hat{k} \quad \checkmark$$

The $\hat{k}$ component captures the same "rotation in the $xy$-plane" as Green's Theorem's $\partial Q/\partial x - \partial P/\partial y$ — which here equals $1-(-1)=2$, exactly as in Session 10 Q3.

---

### Part (b) — Stokes' Theorem ✓

**Stokes' Theorem:**
$$\oint_C \mathbf{F}\cdot d\mathbf{r} = \iint_S (\nabla\times\mathbf{F})\cdot d\mathbf{S}$$

The surface $S$ is the flat disk $x^2+y^2\leq 1$ at $z=1$. Its outward unit normal (consistent with counterclockwise orientation of $C$ when viewed from above) is $\hat{k}$. So:

$$(\nabla\times\mathbf{F})\cdot\hat{k} = 2\hat{k}\cdot\hat{k} = 2$$

This step — dotting the curl with the surface normal — is worth making explicit even when it simplifies nicely, since on a tilted or curved surface the dot product is doing real work.

**Setting up the double integral in polar ✓**

$$\iint_S 2\,dA = \int_0^{2\pi}\int_0^1 2r\,dr\,d\theta
= \int_0^{2\pi}\Big[r^2\Big]_0^1\,d\theta
= \int_0^{2\pi} 1\,d\theta = 2\pi \quad \checkmark$$

---

### Polar coordinate notation — what went wrong and why it matters

You wrote $r^2 = 1 \Rightarrow r = 1$, then $x = \cos\theta$, $y = \sin\theta$,
then set up the integral with $r$ from 0 to 1.

The line $r^2 = 1 \Rightarrow r = 1$ correctly identifies the **upper limit** of the radial integral.
But $x = \cos\theta$, $y = \sin\theta$ is the parametrisation of the **boundary circle**
(where $r$ is fixed at 1) — not the coordinate change for the **interior** of the disk,
where $r$ is a variable.

The correct general polar substitution for an area integral is:
$$x = r\cos\theta, \quad y = r\sin\theta, \quad dA = r\,dr\,d\theta$$
with $r$ running from 0 to the boundary value. Substituting $r=1$ into those formulas
gives $x=\cos\theta$, $y=\sin\theta$ — valid only on the boundary, not throughout the disk.

**Why it didn't matter here.**
The integrand after applying Stokes' Theorem is the constant 2.
Since $x$ and $y$ don't appear in it, the incomplete substitution had no downstream effect.
The integral was assembled correctly with $r$ running from 0 to 1 and the Jacobian factor $r$ included.

**Why it matters in general.**
On any problem where $x$ or $y$ survive into the integrand,
omitting the $r$ factor from the substitution produces a wrong answer.
For example, $\iint_D x^2\,dA$ over the unit disk:

| | Substitution | Integrand | Result |
|---|---|---|---|
| Correct | $x = r\cos\theta$ | $r^2\cos^2\theta \cdot r$ | $\dfrac{\pi}{4}$ |
| Incorrect | $x = \cos\theta$ (boundary only) | $\cos^2\theta \cdot r$ | $\dfrac{\pi}{2}$ — factor of 2 too large |

The error comes from losing the $r^2$ that belongs in $x^2 = r^2\cos^2\theta$.

**The rule to remember: $r$ appears twice in a polar area integral** —
once inside the integrand via $x = r\cos\theta$ (or $y = r\sin\theta$),
and once as the Jacobian in $dA = r\,dr\,d\theta$.
Writing $x = \cos\theta$ silently drops the first occurrence.

### Conceptual explanation ✓

Your explanation is correct: adjacent parallelograms share edges traversed in opposite directions,
so their curl contributions cancel pairwise. Only the outermost boundary edges survive the cancellation,
which is why integrating the curl over any surface with the same boundary gives the same result
as walking that boundary.

To sharpen the phrasing: the conclusion should be stated as *surface independence*, not just
equivalence to the boundary integral. The full statement is:

> For any two surfaces $S_1$ and $S_2$ that share the same oriented boundary $C$,
> $\iint_{S_1}(\nabla\times\mathbf{F})\cdot d\mathbf{S} = \iint_{S_2}(\nabla\times\mathbf{F})\cdot d\mathbf{S}$,
> because both equal $\oint_C \mathbf{F}\cdot d\mathbf{r}$.

Your cancellation argument is also the geometric core of why $\nabla\cdot(\nabla\times\mathbf{F}) = 0$ —
a fact you will use implicitly in Q3.

### Part (a) — Divergence ✓

$$\nabla \cdot \mathbf{F} = \frac{\partial x^3}{\partial x} + \frac{\partial y^3}{\partial y} + \frac{\partial z^3}{\partial z} = 3x^2 + 3y^2 + 3z^2 \quad \checkmark$$

### Part (b) — Divergence Theorem and spherical coordinates ✓

**Divergence Theorem:**
$$\iint_S \mathbf{F}\cdot d\mathbf{S} = \iiint_V \nabla\cdot\mathbf{F}\,dV = \iiint_V (3x^2 + 3y^2 + 3z^2)\,dV \quad \checkmark$$

**Spherical coordinates ✓**

Standard substitution with Jacobian $\rho^2\sin\phi$:

$$x = \rho\sin\phi\cos\theta,\quad y = \rho\sin\phi\sin\theta,\quad z = \rho\cos\phi$$
$$0 \leq \rho \leq 1,\quad 0 \leq \phi \leq \pi,\quad 0 \leq \theta \leq 2\pi \quad \checkmark$$

**Simplification of the integrand ✓ — particularly well done**

You expanded $3x^2 + 3y^2 + 3z^2$ in spherical, then applied the Pythagorean identity twice:

$$3\rho^2\sin^2\phi\cos^2\theta + 3\rho^2\sin^2\phi\sin^2\theta + 3\rho^2\cos^2\phi$$
$$= 3\rho^2\sin^2\phi\underbrace{(\cos^2\theta + \sin^2\theta)}_{=1} + 3\rho^2\cos^2\phi$$
$$= 3\rho^2\underbrace{(\sin^2\phi + \cos^2\phi)}_{=1} = 3\rho^2$$

With the Jacobian $\rho^2\sin\phi$, the integrand becomes $3\rho^4\sin\phi$. ✓

This is exactly the recognition the question was hinting at:
$x^2 + y^2 + z^2 = \rho^2$ in spherical coordinates, so $3(x^2+y^2+z^2) = 3\rho^2$ immediately —
no term-by-term expansion needed. You derived it correctly from scratch, which is the more rigorous route.

**Evaluating the triple integral ✓**

$$\int_0^{2\pi}\int_0^{\pi}\int_0^1 3\rho^4\sin\phi\,d\rho\,d\phi\,d\theta$$

$\rho$ integral: $\displaystyle\int_0^1 3\rho^4\,d\rho = \left[\frac{3\rho^5}{5}\right]_0^1 = \frac{3}{5}$ ✓

$\phi$ integral: $\displaystyle\int_0^{\pi}\frac{3}{5}\sin\phi\,d\phi
= -\frac{3}{5}[\cos\phi]_0^{\pi}
= -\frac{3}{5}(-1-1) = \frac{6}{5}$ ✓

$\theta$ integral: $\displaystyle\int_0^{2\pi}\frac{6}{5}\,d\theta = \frac{12\pi}{5}$ ✓

$$\boxed{\iint_S \mathbf{F}\cdot d\mathbf{S} = \frac{12\pi}{5}}$$

The numerical result $\frac{3}{5}\sin\phi$ is correct either way — purely a presentation point.

### (a) Divergence ✓

Your description is correct and well-phrased. The key ideas are all present:

- Net outward flux per unit volume at a point ✓
- Positive divergence → source (net outflow) ✓
- Negative divergence → sink (net inflow) ✓
- Zero divergence → incompressible / divergence-free ✓
- Physical analogies (sinkhole, fountain) are apt ✓

One small precision: divergence is measured **per unit volume** as the volume shrinks to zero.
The formal definition is:
$$\nabla\cdot\mathbf{F}\,(\mathbf{p}) = \lim_{|V|\to 0} \frac{1}{|V|}\iint_{\partial V} \mathbf{F}\cdot d\mathbf{S}$$
The "per unit volume" is what makes it a local, pointwise quantity rather than a global total flux.
Your description implies this without stating it explicitly — worth keeping in mind for exam answers.

---

### (b) Curl — sign convention is reversed ✗

The paddle-wheel analogy is correct, but you have the sign backwards:

> You wrote: *"a clockwise tendency to rotate is positive and counter-clockwise is negative"*

It is the other way round. **Counter-clockwise rotation gives positive curl; clockwise gives negative curl.**

This follows from the **right-hand rule**: curl your right-hand fingers in the direction of rotation;
your thumb points in the direction of the curl vector. Counter-clockwise rotation (when viewed
from above) → thumb points up → positive $\hat{k}$ component → positive curl.
This is consistent with the $z$-component of curl being $\partial Q/\partial x - \partial P/\partial y$,
which is positive for counter-clockwise rotation (as you computed in Q1: $1 - (-1) = 2 > 0$
for the counter-clockwise field $\mathbf{F} = (-y, x)$).

**Corrected statement:**
> Curl measures the tendency of a vector field to rotate a paddle wheel placed at a point,
> with the paddle axis aligned along the curl vector. Counter-clockwise rotation (by the right-hand rule)
> gives positive curl; clockwise gives negative curl. The magnitude measures the rotation rate.

---

### (c) The unifying principle ✓

Your core idea — interior edges cancel, leaving only the boundary — is exactly right,
and your practical observation (choose whichever form is easier; you can go in both directions)
shows real understanding.

The principle can be stated more precisely as:

> **An integral of a derivative-type operation over a region equals the integral of the
> original quantity over the boundary of that region.**

Each theorem is an instance of this:

| Theorem | Interior (derivative) | Boundary |
|---|---|---|
| Green's | $\iint_D \left(\frac{\partial Q}{\partial x} - \frac{\partial P}{\partial y}\right)dA$ | $\oint_C P\,dx + Q\,dy$ |
| Stokes' | $\iint_S (\nabla\times\mathbf{F})\cdot d\mathbf{S}$ | $\oint_C \mathbf{F}\cdot d\mathbf{r}$ |
| Divergence | $\iiint_V (\nabla\cdot\mathbf{F})\,dV$ | $\iint_S \mathbf{F}\cdot d\mathbf{S}$ |

The abstract version — the **Generalised Stokes' Theorem** — unifies all three in a single formula:
$$\int_{\partial\Omega} \omega = \int_{\Omega} d\omega$$
where $\omega$ is a differential form, $d\omega$ is its exterior derivative, and $\partial\Omega$ is
the oriented boundary of $\Omega$. This is one of the deepest ideas in all of mathematics.
