### Part (b) — Stokes' Theorem ✓

**Stokes' Theorem:**
$$\oint_C \mathbf{F}\cdot d\mathbf{r} = \iint_S (\nabla\times\mathbf{F})\cdot d\mathbf{S}$$

$$(\nabla\times\mathbf{F})\cdot\hat{k} = 2\hat{k}\cdot\hat{k} = 2$$

This step — dotting the curl with the surface normal — is worth making explicit even when it simplifies nicely, since on a tilted or curved surface the dot product is doing real work. (See elaboration below.)

**Setting up the double integral in polar ✓**

$$\iint_S 2\,dA = \int_0^{2\pi}\int_0^1 2r\,dr\,d\theta
= \int_0^{2\pi}\Big[r^2\Big]_0^1\,d\theta
= \int_0^{2\pi} 1\,d\theta = 2\pi \quad \checkmark$$

### Polar coordinate notation — what went wrong and why it matters

You wrote $r^2 = 1 \Rightarrow r = 1$, then $x = \cos\theta$, $y = \sin\theta$,
then set up the integral with $r$ from 0 to 1.

The line $r^2 = 1 \Rightarrow r = 1$ correctly identifies the **upper limit** of the radial integral.
But $x = \cos\theta$, $y = \sin\theta$ is the parametrisation of the **boundary circle**
(where $r$ is fixed at 1) — not the coordinate change for the **interior** of the disk,
where $r$ is a variable.

The correct general polar substitution for an area integral is:
$$x = r\cos\theta, \quad y = r\sin\theta, \quad dA = r\,dr\,d\theta$$
with $r$ running from 0 to the boundary value. Substituting $r=1$ into those formulas
gives $x=\cos\theta$, $y=\sin\theta$ — valid only on the boundary, not throughout the disk.

**Why it didn't matter here.**
The integrand after applying Stokes' Theorem is the constant 2.
Since $x$ and $y$ don't appear in it, the incomplete substitution had no downstream effect.
The integral was assembled correctly with $r$ running from 0 to 1 and the Jacobian factor $r$ included.

**Why it matters in general.**
On any problem where $x$ or $y$ survive into the integrand,
omitting the $r$ factor from the substitution produces a wrong answer.
For example, $\iint_D x^2\,dA$ over the unit disk:

| | Substitution | Integrand | Result |
|---|---|---|---|
| Correct | $x = r\cos\theta$ | $r^2\cos^2\theta \cdot r$ | $\dfrac{\pi}{4}$ |
| Incorrect | $x = \cos\theta$ (boundary only) | $\cos^2\theta \cdot r$ | $\dfrac{\pi}{2}$ — factor of 2 too large |

The error comes from losing the $r^2$ that belongs in $x^2 = r^2\cos^2\theta$.

**The rule to remember: $r$ appears twice in a polar area integral** —
once inside the integrand via $x = r\cos\theta$ (or $y = r\sin\theta$),
and once as the Jacobian in $dA = r\,dr\,d\theta$.
Writing $x = \cos\theta$ silently drops the first occurrence.

### Elaboration — What the dot product is doing on tilted and curved surfaces

In Q1 the dot product $(\nabla\times\mathbf{F})\cdot\hat{n}$ reduced to the constant 2 because the flat disk has the same normal $\hat{k}$ everywhere, and the curl $2\hat{k}$ happens to point in exactly that direction. In that situation the dot product does nothing interesting. The remark was pointing forward to cases where it does real work.

---

#### What the dot product is measuring

The curl vector $\nabla\times\mathbf{F}$ points along the axis of the local rotation in the field — like the axle of a spinning wheel. The surface normal $\hat{n}$ points perpendicular to the surface — like the direction you would push a corkscrew through the fabric.

The dot product $(\nabla\times\mathbf{F})\cdot\hat{n}$ measures how much of the spin axis threads through the surface:

- **Spin axis parallel to normal** → corkscrew goes straight through → dot product = full magnitude → maximum contribution to the flux integral.
- **Spin axis perpendicular to normal** → corkscrew is parallel to the fabric → dot product = 0 → no contribution.
- **Spin axis at an angle** → partial threading → dot product somewhere between 0 and the full magnitude.

In Q1, spin axis = $2\hat{k}$ and surface normal = $\hat{k}$: perfectly aligned, so the full magnitude 2 contributes everywhere. The dot product is trivial.

---

#### Where the formula $-Pg_x - Qg_y + R$ comes from

For a surface parameterised as $z = g(x,y)$, every point can be written as $\mathbf{r}(x,y) = (x,\, y,\, g(x,y))$. The two tangent vectors to the surface are the partial derivatives of this position vector:

$$\mathbf{r}_x = (1,\, 0,\, g_x), \qquad \mathbf{r}_y = (0,\, 1,\, g_y)$$

These are the sides of the infinitesimal parallelogram that tiles the surface at each point. Their cross product gives the surface normal, already scaled by the correct area factor:

$$\mathbf{r}_x \times \mathbf{r}_y = (-g_x,\, -g_y,\, 1)$$

So $d\mathbf{S} = (\mathbf{r}_x \times \mathbf{r}_y)\,dA = (-g_x, -g_y, 1)\,dA$, and the Stokes' integrand is:

$$(\nabla\times\mathbf{F})\cdot d\mathbf{S} = (P,\,Q,\,R)\cdot(-g_x,\,-g_y,\,1)\,dA = (-Pg_x - Qg_y + R)\,dA$$

where $(P, Q, R) = \nabla\times\mathbf{F}$. **The formula is not a separate recipe — it is the dot product, derived directly from the parallelogram geometry of the surface.** The three terms correspond to: how much the $x$-tilt of the surface reduces the $x$-component of the curl's threading, how much the $y$-tilt does the same, and the full $z$-component passing straight through.

**For the flat disk at $z = 1$:** here $g(x,y) = 1$, so $g_x = g_y = 0$. Substituting:

$$-P(0) - Q(0) + R = R = (\nabla\times\mathbf{F})_z = 2$$

The formula immediately reduces to just the $z$-component of the curl. This is the clean, explicit route: identify $g_x$ and $g_y$ for the surface, substitute into the formula, read off the integrand. No separate reasoning about the normal direction is required — the formula encodes it automatically.

The formula does real work when the surface is tilted or curved and $g_x$, $g_y$ are nonzero — as in Q2, where $g_x = -x/\sqrt{1-x^2-y^2}$ and $g_y = -y/\sqrt{1-x^2-y^2}$. There the formula carries all the geometric information about how the hemisphere's tilt varies relative to the curl direction at each point.

---

#### Example: the same boundary, a tilted surface

From Q2 you know Stokes' Theorem is **surface-independent**: any surface bounded by the same curve $C$ gives the same answer. So take a tilted surface instead of the flat disk — say the planar cap cut by $z = 1 + y$ over the unit disk.

Here $g(x,y) = 1 + y$, so $g_x = 0$, $g_y = 1$. Applying the formula with $\nabla\times\mathbf{F} = (0, 0, 2)$:

$$-P(0) - Q(1) + R = -(0)(0) - (0)(1) + 2 = 2$$

Wait — the integrand is still 2? Yes, because the curl has no $x$- or $y$-components, so the tilt terms vanish regardless. To see the tilt making a visible difference, we would need a curl with nonzero $x$- or $y$-components. But the area of the tilted surface is $\sqrt{2}$ times that of the flat disk, and the unnormalised normal $(-g_x, -g_y, 1) = (0, -1, 1)$ has magnitude $\sqrt{2}$, so the integral over the tilted surface correctly equals $2 \times \sqrt{2} \times \frac{\pi}{\sqrt{2}} = 2\pi$, matching the flat disk. The dot product and area element compensate for each other, preserving surface independence.

---

#### Example: a curved surface (hemisphere)

On the hemisphere in Q2, the outward normal changes direction at every point. In spherical coordinates at angles $(\phi,\theta)$:

$$\hat{n}(\phi,\theta) = (\sin\phi\cos\theta,\;\sin\phi\sin\theta,\;\cos\phi)$$

With $\nabla\times\mathbf{F} = (0,0,2)$, the dot product at each point is:

$$(0,0,2)\cdot(\sin\phi\cos\theta,\;\sin\phi\sin\theta,\;\cos\phi) = 2\cos\phi$$

This is **not constant**: it equals 2 at the top of the dome ($\phi=0$, where the surface faces straight up like the flat disk) and drops to zero at the equator ($\phi=\pi/2$, where the surface faces horizontally and the spin axis is perpendicular to it). The integral then weights each surface element by how much of the spin axis threads through it at that location.

The varying integrand integrates to the same $2\pi$ — as it must, by surface independence.

---

#### Why this was easy in Q1 and harder in general

Q1 used the simplest possible surface: flat, parallel to the $xy$-plane, with $g_x = g_y = 0$. The formula $-Pg_x - Qg_y + R$ reduced to $R$ immediately. When you encounter Stokes' Theorem problems on tilted planes or curved surfaces, computing $g_x$ and $g_y$ and substituting into the formula is the step that carries all the geometric information — it cannot be skipped.

### (c) The unifying principle ✓

Your core idea — interior edges cancel, leaving only the boundary — is exactly right,
and your practical observation (choose whichever form is easier; you can go in both directions)
shows real understanding.

The principle can be stated more precisely as:

> **An integral of a derivative-type operation over a region equals the integral of the
> original quantity over the boundary of that region.**

Each theorem is an instance of this:

| Theorem | Interior (derivative) | Boundary |
|---|---|---|
| Green's | $\iint_D \left(\frac{\partial Q}{\partial x} - \frac{\partial P}{\partial y}\right)dA$ | $\oint_C P\,dx + Q\,dy$ |
| Stokes' | $\iint_S (\nabla\times\mathbf{F})\cdot d\mathbf{S}$ | $\oint_C \mathbf{F}\cdot d\mathbf{r}$ |
| Divergence | $\iiint_V (\nabla\cdot\mathbf{F})\,dV$ | $\iint_S \mathbf{F}\cdot d\mathbf{S}$ |

The abstract version — the **Generalised Stokes' Theorem** — unifies all three in a single formula:
$$\int_{\partial\Omega} \omega = \int_{\Omega} d\omega$$
where $\omega$ is a differential form, $d\omega$ is its exterior derivative, and $\partial\Omega$ is
the oriented boundary of $\Omega$. This is one of the deepest ideas in all of mathematics.
(See elaboration below.)

### Elaboration — Reading the notation and why the theorem is genuinely deep

You raised two questions: how to read the notation, and why this deserves to be called one of the deepest ideas in mathematics when the underlying argument (interior edges cancel) doesn't seem that profound. Both are fair. The notation first, then the depth.

---

#### A concrete dictionary for $\int_{\partial\Omega}\omega = \int_{\Omega}d\omega$

**$\Omega$ and $\partial\Omega$ — the domain and its boundary**

| Theorem | $\Omega$ (what you integrate over) | $\partial\Omega$ (its boundary) |
|---|---|---|
| Fundamental Theorem of Calculus | interval $[a,b]$ | two endpoints: $\{b\}$ (positive) and $\{a\}$ (negative) |
| Green's | region $D$ in the plane | boundary curve $C$, counterclockwise |
| Stokes' | surface $S$ in 3D | boundary curve $C$ of that surface |
| Divergence | volume $V$ | closed boundary surface $\partial V$, outward normal |

**$\omega$ — the differential form (what appears after the integral sign)**

A differential form is the object you integrate. You have been writing these throughout the course:

- Along a curve: $\omega = P\,dx + Q\,dy + R\,dz$ — this is $\mathbf{F}\cdot d\mathbf{r}$, a **1-form**.
- Over a surface: $\omega = P\,dy\wedge dz + Q\,dz\wedge dx + R\,dx\wedge dy$ — this is $\mathbf{F}\cdot d\mathbf{S}$, a **2-form**.
- Over a volume: $\omega = f\,dx\wedge dy\wedge dz$ — this is $f\,dV$, a **3-form**.

The wedge $\wedge$ is the oriented area (or volume) element. It keeps track of orientation by swapping sign when you swap the order: $dy\wedge dx = -dx\wedge dy$. This is the algebraic encoding of the fact that reversing a curve's direction negates the integral.

**$d\omega$ — the exterior derivative (all your derivative operators, unified)**

The operation $d$ is a single generalised derivative. Applied to each type of form it produces exactly the operators you already know:

| Applied to | $d$ produces | You recognise it as |
|---|---|---|
| A function $f$ | $\displaystyle\frac{\partial f}{\partial x}dx + \frac{\partial f}{\partial y}dy + \frac{\partial f}{\partial z}dz$ | Gradient $\nabla f$ (written as a 1-form) |
| A 1-form $P\,dx + Q\,dy$ | $\displaystyle\left(\frac{\partial Q}{\partial x} - \frac{\partial P}{\partial y}\right)dx\wedge dy$ | The curl $z$-component / Green's integrand |
| A 1-form $\mathbf{F}\cdot d\mathbf{r}$ in 3D | $(\nabla\times\mathbf{F})\cdot d\mathbf{S}$ | Curl (Stokes' integrand) |
| A 2-form $\mathbf{F}\cdot d\mathbf{S}$ | $\displaystyle\left(\frac{\partial P}{\partial x}+\frac{\partial Q}{\partial y}+\frac{\partial R}{\partial z}\right)dx\wedge dy\wedge dz$ | Divergence (Divergence Theorem integrand) |

**The full dictionary**

Substituting these into $\int_{\partial\Omega}\omega = \int_\Omega d\omega$ gives back the four theorems:

| Theorem | $\int_{\partial\Omega}\omega$ | $=$ | $\int_\Omega d\omega$ |
|---|---|---|---|
| FTC | $f(b) - f(a)$ | $=$ | $\displaystyle\int_a^b f'(x)\,dx$ |
| Green's | $\displaystyle\oint_C P\,dx+Q\,dy$ | $=$ | $\displaystyle\iint_D \left(\frac{\partial Q}{\partial x}-\frac{\partial P}{\partial y}\right)dA$ |
| Stokes' | $\displaystyle\oint_C \mathbf{F}\cdot d\mathbf{r}$ | $=$ | $\displaystyle\iint_S (\nabla\times\mathbf{F})\cdot d\mathbf{S}$ |
| Divergence | $\displaystyle\iint_S \mathbf{F}\cdot d\mathbf{S}$ | $=$ | $\displaystyle\iiint_V (\nabla\cdot\mathbf{F})\,dV$ |

---

#### Why this is genuinely profound

You raised a fair challenge: the interior-cancellation argument doesn't feel particularly deep on its own. That instinct is correct — the cancellation is a *proof technique*, not the profound content. The depth lies elsewhere — in several places at once.

**1. One theorem replaces four, and extends to any dimension.**
Calculus courses teach the FTC, Green's, Stokes', and Divergence theorems as separate results with separate proofs. The Generalised Stokes' Theorem is a single equation that contains all four — and continues into 4, 5, and $n$ dimensions without any modification. There is, in a precise sense, only one theorem of integration in all of mathematics.

**2. The FTC is a special case, not a separate idea.**
You learned the Fundamental Theorem of Calculus in first-year: $\int_a^b f'(x)\,dx = f(b) - f(a)$.
In the generalised framework, the interval $[a,b]$ is a 1-dimensional $\Omega$, and its boundary $\partial[a,b]$ is the two endpoints $\{b\} - \{a\}$ (with orientation). The FTC is $\int_{\partial\Omega}\omega = \int_\Omega d\omega$ with $\Omega = [a,b]$. The theorem you first met in first-year calculus and the Divergence Theorem you used this session are instances of the same single equation.

**3. Gradient, curl, and divergence are the same operation.**
These three operators look unrelated — different input types, different output types, different formulas. Under the exterior derivative $d$, they are all one operation applied to forms of different degree. The formal diversity of vector calculus is an artefact of working in 3D; in the language of forms, it collapses to a single symbol. (See forms elaboration below for a step-by-step derivation.)

**4. The identity $d^2 = 0$ is where the real algebra lives.**
Applying $d$ twice always gives zero. Concretely:

$$\nabla\times(\nabla f) = \mathbf{0} \qquad \text{(curl of a gradient is zero)}$$
$$\nabla\cdot(\nabla\times\mathbf{F}) = 0 \qquad \text{(divergence of a curl is zero)}$$

These look like two separate vector-calculus identities. They are both the single algebraic fact $d^2 = 0$. The reason it holds is Clairaut's theorem: mixed partial derivatives commute ($\partial^2 f / \partial x\partial y = \partial^2 f / \partial y\partial x$), so applying $d$ twice always produces pairs of terms that cancel. A single fact from multivariable calculus — one you proved implicitly in every partial-derivative calculation — underpins the entire algebraic structure.

**5. It connects the shape of a region to equations inside it.**
If $\nabla\cdot\mathbf{F} = 0$ everywhere inside $V$ (no sources), then $\iint_{\partial V}\mathbf{F}\cdot d\mathbf{S} = 0$ (no net flux through any closed surface). Information about local behaviour in the interior propagates to the boundary, and vice versa. This boundary-to-interior passage is the mathematical engine behind every conservation law in physics — conservation of mass, charge, energy, momentum.

**6. Modern physics is written in this language.**
Maxwell's equations — all four of them — reduce to two equations in differential-form notation: $dF = 0$ and $d{\star}F = J$, where $F$ is the electromagnetic field 2-form. The integral versions of Maxwell's equations (Gauss's law, Faraday's law, Ampère's law) follow directly from the Generalised Stokes' Theorem applied to these. General relativity and quantum field theory use the same machinery. The theorem is not a cosmetic reformulation — it is the actual language in which the fundamental laws of nature are expressed.

---

#### Summary of where the depth is

The interior-cancellation argument explains **why** $\int_{\partial\Omega}\omega = \int_\Omega d\omega$ is true — it is the proof. The depth is that:

- This single equation unifies all of integration theory across every dimension.
- Gradient, curl, and divergence are one operation; the FTC is the $n=1$ case.
- The algebraic identity $d^2 = 0$ (from Clairaut's theorem) is what makes the structure coherent.
- Conservation laws in physics are consequences of this identity applied to the laws of nature.

The cancellation gives you the proof. The profundity is that one equation governs integration from the number line to the fabric of spacetime.

### Elaboration — Differential forms and why gradient, curl, and divergence are one operation

This section builds the idea from scratch. No prior knowledge of forms is assumed.

---

#### What is a differential form?

A differential form is the object that appears after an integral sign. You have been writing them throughout this course without using the name.

**0-forms** are ordinary functions $f(x,y,z)$. You evaluate a 0-form at a point.

**1-forms** are expressions like $P\,dx + Q\,dy + R\,dz$. These are what you put after $\int$ when integrating along a curve. Every time you wrote $\int_C \mathbf{F}\cdot d\mathbf{r} = \int_C P\,dx + Q\,dy + R\,dz$, the object being integrated was a 1-form.

**2-forms** are expressions like $P\,dy\wedge dz + Q\,dz\wedge dx + R\,dx\wedge dy$. These are what you put after $\iint$ when integrating over a surface. The expression $\mathbf{F}\cdot d\mathbf{S}$ is a 2-form. The wedge product $\wedge$ is antisymmetric: $dy\wedge dx = -dx\wedge dy$ and $dx\wedge dx = 0$. This sign rule is the algebraic way of encoding orientation — reversing the order of traversal negates the integral.

**3-forms** are expressions like $f\,dx\wedge dy\wedge dz = f\,dV$. These are what you integrate over volumes.

| Form degree | Object | Integrated over |
|---|---|---|
| 0-form | function $f$ | a point (evaluate) |
| 1-form | $P\,dx + Q\,dy + R\,dz$ | a curve |
| 2-form | $P\,dy\wedge dz + Q\,dz\wedge dx + R\,dx\wedge dy$ | a surface |
| 3-form | $f\,dx\wedge dy\wedge dz$ | a volume |

---

#### Basis elements

Every form is built from terms of the shape **(coefficient function) $\times$ (basis element)**. The basis elements are the standard building blocks for each degree — analogous to $\hat{\imath}$, $\hat{\jmath}$, $\hat{k}$ for vector fields, but for integrands:

| Degree | Basis elements | What they represent |
|---|---|---|
| 0-form | $1$ (just a number) | a function has no differential attached |
| 1-form | $dx,\; dy,\; dz$ | signed length element in each direction |
| 2-form | $dy\wedge dz,\; dz\wedge dx,\; dx\wedge dy$ | signed area element in each coordinate plane |
| 3-form | $dx\wedge dy\wedge dz$ | signed volume element |

So the 1-form $P\,dx + Q\,dy + R\,dz$ is a linear combination of the basis elements $dx$, $dy$, $dz$ with coefficient functions $P$, $Q$, $R$ — exactly the same structure as a vector field $P\hat{\imath} + Q\hat{\jmath} + R\hat{k}$, but for something you integrate along curves instead of something you evaluate at points.

---

#### The exterior derivative $d$ — the rule and a mini-example

The rule for applying $d$ to a single term is:

> **Step 1.** Differentiate the coefficient function, obtaining a new 1-form.
> **Step 2.** Wedge that result onto the existing basis element.
> **Step 3.** Expand and simplify using $dx\wedge dx = 0$ and $dy\wedge dx = -dx\wedge dy$.

The full $d$ of an entire form is the sum of this applied term by term.

**Mini-example: $d$ applied to the single term $P\,dx$.**

The coefficient is $P$ and the existing basis element is $dx$.

**Step 1.** Differentiate $P$:
$$dP = \frac{\partial P}{\partial x}\,dx + \frac{\partial P}{\partial y}\,dy + \frac{\partial P}{\partial z}\,dz$$

**Step 2.** Wedge with the existing basis element $dx$:
$$d(P\,dx) = dP \wedge dx = \left(\frac{\partial P}{\partial x}\,dx + \frac{\partial P}{\partial y}\,dy + \frac{\partial P}{\partial z}\,dz\right)\wedge dx$$

**Step 3.** Expand the wedge product term by term:
$$= \frac{\partial P}{\partial x}\underbrace{dx\wedge dx}_{=\;0}
\;+\; \frac{\partial P}{\partial y}\underbrace{dy\wedge dx}_{=\;-\,dx\wedge dy}
\;+\; \frac{\partial P}{\partial z}\underbrace{dz\wedge dx}_{\text{(leave as-is)}}$$

$$= -\frac{\partial P}{\partial y}\,dx\wedge dy \;+\; \frac{\partial P}{\partial z}\,dz\wedge dx$$

The $dx\wedge dx$ term vanishes because you cannot have a nonzero oriented area with both edges pointing in the same direction. The $dy\wedge dx$ term flips sign because swapping the two directions reverses orientation (clockwise vs counter-clockwise). These two rules — together with linearity — are all that is needed to carry out any exterior derivative calculation.

The gradient, curl, and divergence derivations below each apply exactly this three-step process, summed over all terms in the form.

---

#### $d$ on a 0-form → gradient

Given a function (0-form) $f$, its only "basis element" is $1$ (no differential attached), so Step 2 just attaches $d$ directly:

$$d(f) = \frac{\partial f}{\partial x}\,dx + \frac{\partial f}{\partial y}\,dy + \frac{\partial f}{\partial z}\,dz$$

The coefficients $\left(\frac{\partial f}{\partial x},\, \frac{\partial f}{\partial y},\, \frac{\partial f}{\partial z}\right)$ are exactly the gradient vector $\nabla f$.

**Example.** Take $f = x^2 + yz$:
$$d(f) = 2x\,dx + z\,dy + y\,dz$$
The coefficient vector $(2x, z, y)$ is $\nabla f$. ✓

---

#### $d$ on a 1-form → curl

Given a 1-form $\omega = P\,dx + Q\,dy + R\,dz$, apply the three-step rule to each of the three terms and sum:

$$d(\omega) = d(P\,dx) + d(Q\,dy) + d(R\,dz)$$

From the mini-example above:
$$d(P\,dx) = -\frac{\partial P}{\partial y}\,dx\wedge dy + \frac{\partial P}{\partial z}\,dz\wedge dx$$

The same process on $d(Q\,dy)$ gives $\frac{\partial Q}{\partial x}\,dx\wedge dy - \frac{\partial Q}{\partial z}\,dy\wedge dz$, and on $d(R\,dz)$ gives $\frac{\partial R}{\partial y}\,dy\wedge dz - \frac{\partial R}{\partial x}\,dz\wedge dx$. Collecting by basis element:

$$d(\omega) =
\left(\frac{\partial R}{\partial y} - \frac{\partial Q}{\partial z}\right)dy\wedge dz
+ \left(\frac{\partial P}{\partial z} - \frac{\partial R}{\partial x}\right)dz\wedge dx
+ \left(\frac{\partial Q}{\partial x} - \frac{\partial P}{\partial y}\right)dx\wedge dy$$

The three coefficients are exactly the three components of $\nabla\times\mathbf{F}$.

**Example from Q1.** $\mathbf{F} = (-y, x, z^2)$ corresponds to $\omega = -y\,dx + x\,dy + z^2\,dz$.

Applying the rule to each term:
$$d(-y\,dx) = d(-y)\wedge dx = (-dy)\wedge dx = dx\wedge dy$$
$$d(x\,dy) = d(x)\wedge dy = (dx)\wedge dy = dx\wedge dy$$
$$d(z^2\,dz) = d(z^2)\wedge dz = (2z\,dz)\wedge dz = 0$$

Summing: $d(\omega) = dx\wedge dy + dx\wedge dy + 0 = 2\,dx\wedge dy$

This 2-form has coefficient 2 on the $dx\wedge dy$ basis element and zero on the others, corresponding to curl vector $(0, 0, 2) = 2\hat{k}$. ✓ This is exactly what the determinant formula gave in Q1(a).

---

#### $d$ on a 2-form → divergence

Given $\omega = P\,dy\wedge dz + Q\,dz\wedge dx + R\,dx\wedge dy$, apply the rule to each term. Each coefficient differentiates to a full $dP$, $dQ$, $dR$, then wedges with the existing 2-form basis element. All terms with a repeated differential vanish, leaving only the terms that complete the triple $dx\wedge dy\wedge dz$:

$$d(\omega) = \left(\frac{\partial P}{\partial x} + \frac{\partial Q}{\partial y} + \frac{\partial R}{\partial z}\right)dx\wedge dy\wedge dz = (\nabla\cdot\mathbf{F})\,dV$$

**Example from Q3.** $\mathbf{F} = (x^3, y^3, z^3)$ corresponds to $\omega = x^3\,dy\wedge dz + y^3\,dz\wedge dx + z^3\,dx\wedge dy$:

$$d(\omega) = \left(\frac{\partial x^3}{\partial x} + \frac{\partial y^3}{\partial y} + \frac{\partial z^3}{\partial z}\right)dx\wedge dy\wedge dz = (3x^2 + 3y^2 + 3z^2)\,dV$$

This is exactly the divergence from Q3(a). ✓

---

#### The three operators as one sequence

The exterior derivative $d$ maps along the chain:

$$\underbrace{f}_{\text{0-form}}
\xrightarrow{\;d\;}
\underbrace{\omega_1}_{\text{1-form}}
\xrightarrow{\;d\;}
\underbrace{\omega_2}_{\text{2-form}}
\xrightarrow{\;d\;}
\underbrace{\omega_3}_{\text{3-form}}
\xrightarrow{\;d\;} 0$$

Translating into 3D vector language:

$$\underbrace{\text{scalar}\; f}_{\text{function}}
\xrightarrow{\;\nabla\;}
\underbrace{\nabla f}_{\text{vector field}}
\xrightarrow{\;\nabla\times\;}
\underbrace{\nabla\times\mathbf{F}}_{\text{vector field}}
\xrightarrow{\;\nabla\cdot\;}
\underbrace{\nabla\cdot\mathbf{F}}_{\text{scalar}}
\xrightarrow{} 0$$

Gradient, curl, and divergence are not three different operations. They are the same operation $d$, applied to objects of dimension 0, 1, and 2 respectively.

The final $\to 0$ is $d^2 = 0$, which gives both vector-calculus identities at once:

$$d(d(f)) = 0 \quad\Leftrightarrow\quad \nabla\times(\nabla f) = \mathbf{0}$$
$$d(d(\omega_1)) = 0 \quad\Leftrightarrow\quad \nabla\cdot(\nabla\times\mathbf{F}) = 0$$

Both follow from the same cause: Clairaut's theorem ($\partial^2 f/\partial x\partial y = \partial^2 f/\partial y\partial x$) means every term produced by $d^2$ appears twice with opposite signs and cancels.

#### Why this is an artefact of 3D

In $n$-dimensional space, a $k$-form has $\binom{n}{k}$ independent basis elements.

| | 0-forms | 1-forms | 2-forms | 3-forms | 4-forms |
|---|---|---|---|---|---|
| 2D | 1 | 2 | 1 | — | — |
| **3D** | **1** | **3** | **3** | **1** | — |
| 4D | 1 | 4 | 6 | 4 | 1 |

In 3D, $\binom{3}{1} = \binom{3}{2} = 3$: 1-forms and 2-forms both have exactly 3 components. This numerical coincidence is what allows us to represent both as vector fields and write curl as a vector-to-vector operation. The curl of a vector field outputs another vector field only because of this accident of dimension.

In 4D, 2-forms have 6 independent components — they cannot be represented as a 4-component vector field. There is no "curl" in the 4D vector-calculus sense. The exterior derivative $d$ works perfectly in any dimension; it is only in 3D that it can be dressed up as vector operations on all three steps.

**Gradient, curl, and divergence are the 3D costume that $d$ wears in three dimensions.** Underneath, there is one operation.
