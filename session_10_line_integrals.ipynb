## Q1 — Line integral of a scalar field

Evaluate $\displaystyle\int_C (x + y^2) \, ds$ where $C$ is the line segment from $(0, 0)$ to $(1, 2)$.

Parametrise the curve, express $ds$ in terms of the parameter, and evaluate.

**Your answer:**
$$
\begin{gather*}
& x=t \\
& y=2t \\
& r(t) = t\hat{i} + 2t\hat{j} \\
& r^{\prime}(t) = 1\hat{i} + 2\hat{j} \\
& \left|r^{\prime}(t)\right|=\sqrt{(1)^2+(2)^2}=\sqrt{5} \\
& \int_{t=0}^{t=1}\left(t+(2 t)^2\right) \sqrt{5} dt \\
& \sqrt{5} \int_{t=0}^{t=1}\left(t+4 t^2\right) dt \\
& \sqrt{5}\left[\frac{t^2}{2}+\frac{4 t^3}{3}\right]_{t=0}^{t=1} \\
& =\sqrt{5}\left[\frac{1}{2} + \frac{4}{3}\right] \\
& =\sqrt{5} \frac{11}{6} 
\end{gather*}
$$
