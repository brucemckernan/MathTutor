## Q1 — Line integral of a scalar field

Evaluate $\displaystyle\int_C (x + y^2) \, ds$ where $C$ is the line segment from $(0, 0)$ to $(1, 2)$.

Parametrise the curve, express $ds$ in terms of the parameter, and evaluate.

**Your answer:**
$$
\begin{gather*}
& x=t \\
& y=2t \\
& r(t) = t\hat{i} + 2t\hat{j} \\
& r^{\prime}(t) = 1\hat{i} + 2\hat{j} \\
& \left|r^{\prime}(t)\right|=\sqrt{(1)^2+(2)^2}=\sqrt{5} \\
& \int_{t=0}^{t=1}\left(t+(2 t)^2\right) \sqrt{5} dt \\
& \sqrt{5} \int_{t=0}^{t=1}\left(t+4 t^2\right) dt \\
& \sqrt{5}\left[\frac{t^2}{2}+\frac{4 t^3}{3}\right]_{t=0}^{t=1} \\
& =\sqrt{5}\left[\frac{1}{2} + \frac{4}{3}\right] \\
& =\sqrt{5} \frac{11}{6} 
\end{gather*}
$$


## Q2 — Conservative fields

Let $\mathbf{F}(x, y) = (2xy + y^3)\mathbf{i} + (x^2 + 3xy^2)\mathbf{j}$.

(a) Show that $\mathbf{F}$ is conservative by checking the appropriate condition.

(b) Find a potential function $\varphi$ such that $\nabla \varphi = \mathbf{F}$.

(c) Use the Fundamental Theorem for Line Integrals to evaluate $\int_C \mathbf{F} \cdot d\mathbf{r}$ where $C$ is any path from $(0, 0)$ to $(1, 2)$.

**Your answer:**
\begin{gather*}
& F(x, y)=\left(2 x y+y^3\right) \hat{i}+\left(x^2+3 x y^2\right) \hat{j} \\
& \frac{\partial}{\partial y}\left(2 x y+y^3\right)=2 x+3 y^2 \\
& \frac{\partial}{\partial x}\left(x^2+3 x y^2\right)=2 x+3 y^2 \\
\end{gather*}
There is zero curl therefore the Vector Field is conservative (i.e. the parial derivatives are equal). I start by integrating with respect to x in order to get a definition of the potential function f where y is left undefined at this point.
\begin{gather*}
& \int 2 x y+y^3 d x \\
& f(x, y)=x^2 y+x y^3+g(y) \\
\end{gather*}

I then derive the resulting partial definition of f(x,y) with respect to y, which gives me a definition of f'(x,y) that I can match to the definition of df/dy that I have in the original Vector Field definition. By setting them equal I can solve for g'(y).

\begin{gather*}
& \frac{\partial}{\partial y}\left(x^2 y+x y^3+g(y)\right)=x^2+3 x y^2+g^{\prime}(y) \\
& x^2+3 x y^2+g^{\prime}(y)=x^2+3 x y^2 \\
& g^{\prime}(y)=0, g(y)=C \\
& f(x, y)=x^2 y+x y^3+C \\
\end{gather*}

In this case there is no extra info required about df/dy i.e. the initial definition of f(x,y) contained all the information about both x and y. Therefore g(y) solves to the constant C. I can therefore compose the potential function f, and use the fundamental theorem for line integrals and tha fact that a conservative vector field has independence of path to just plug in the start and end points:

\begin{gather*}
& \int F . d r=\int \nabla f . d r=f(b)-f(a) \\
& f(x, y)=x^2 y+x y^3+C \\
& f(0,0)=C \\
& f(1,2)=2+8+C=10+C \\
& f(b)-f(a)=10+C-C=10 \\
\end{gather*}


## Q3 — Non-conservative field

Let $\mathbf{F}(x,y) = (-y, x)$. Show that $\mathbf{F}$ is not conservative, then evaluate $\oint_C \mathbf{F} \cdot d\mathbf{r}$ directly where $C$ is the unit circle traversed counterclockwise.

**Your answer:**
For a simple closed curve in a non conservative vector field I would nrtomally use Greens theorem. However the question asks for the solution to be calculated directly :

\begin{gather*}
& F(x, y)=(-y, x) \\
& P=-y, Q=x \\
& \frac{\partial P}{\partial y}=-1, \frac{\partial Q}{\partial x}=1 \\
\end{gather*}

The partial derivatives of F are unequal and therefore the vector field is not conservative. As I have a circular path (not a region!) with a radius of 1, I can convert the variables to a parameterised form based on t using cosine and sine:

\begin{gather*}
& x=\cos t \\
& y=\sin t \\
& r(t)=\cos t \hat{i}+\sin t \hat{j} \\
& r^{\prime}(t)=-\sin t \hat{i}+\cos t \hat{j} \\
\end{gather*}

I convert the vector field to the same definitions of x and y

\begin{gather*}
& F(x(t), y(t))=(-\sin t, \cos t) \\
& \oint F \cdot dr=\int_{t=0}^{t=2 \pi}(-\sin t, \cos t) \cdot-\sin t \hat{i}+\cos t \hat{j} dt \\
& =\int_{t=0}^{t=2 \pi}-\sin ^2 t+\cos ^2 t \quad dt
\end{gather*}

I can drop the leading minus on sin as it is squared and therefore always positive. I then use a trig identity to convert the result to 1:

\begin{gather*}
& =\int_{t=0}^{t=2 \pi} \sin ^2 t+\cos ^2 t d t \\
& =\int_{t=0}^{t=2 \pi} 1 d t \\
& =[t]_{t=0}^{t=2 \pi} \\
& =2 \pi-0=2 \pi
\end{gather*}

## Q4 — Green's Theorem

Use Green's Theorem to evaluate $\oint_C (y^2 \, dx + x \, dy)$ where $C$ is the boundary of the region bounded by $y = x$ and $y = x^2$, traversed counterclockwise.

State Green's Theorem clearly, identify $P$ and $Q$, and set up the resulting double integral.

**Your answer:**
I begin by making sure I have a non conservative vector field, which I do

\begin{gather*}
& \oint_C\left(y^2 d x+x d y\right) \quad y=x, y=x^2 \\
& P=y^2, Q=x \\
& \frac{\partial P}{\partial y}=2 y, \frac{\partial Q}{\partial x}=1 \\
\end{gather*}

I then do a bit of work on the region formed by the two curves to make sure it is a simple closed curve (it is and they intersect at points (0,0) and (1,1) ) and I use the point x=0.5 to show that y=x^2 is on the bottom:

\begin{gather*}
& x=x^2 \longrightarrow(0,0),(1,1) \\
& \text { Let } x=0.5 \\
& y=x=0.5 \\
& y=x^2=0.25 \\
\end{gather*}

I then apply Greens Theorem to convert the traversal of a region boundary into a double integral of the Curl of the region bounded by the curves:

\begin{gather*}
& \oint_C\left(y^2 \partial x+x \partial y\right)=\iint_D \frac{\partial}{\partial x} x-\frac{\partial}{\partial y} y^2 d A \\
& =\iint_D 1-2 y d A \\
\end{gather*}

As I have equations already solved for y I choose y-simple and integrate in the y direction first, setting up the boundaries and order of integration in dydx order

\begin{gather*}
& =\int_{x=0}^{x=1} \int_{y=x^2}^{y=x} 1-2 y d y d x \\
& =\int_{x=0}^{x=1}\left[y-y^2\right]_{y=x^2}^{y=x} d x \\
& =\int_{x=0}^{x=1}\left(x-x^2\right)-\left(x^2-\left(x^2\right)^2\right) d x \\
& =\int_{x=0}^{x=1}\left(x-x^2\right)-\left(x^2-x^4\right) d x \\
& =\int_{x=0}^{x=1} x-x^2-x^2+x^4 d x \\
& =\int_{x=0}^{x=1} x d x \\
& =\left[\frac{x^2}{2}\right]_{x=0}^{x=1} \\
& =\frac{1}{2}
\end{gather*}
