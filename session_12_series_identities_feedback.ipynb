### Q1(b)

**Method correct; arithmetic error at the final step.**

Your approach was exactly right:
$$\sum_{n=2}^\infty 4\left(\frac{1}{5}\right)^n = 4\left(\frac{1}{5}\right)^2 \cdot \sum_{n=0}^\infty \left(\frac{1}{5}\right)^n = \frac{4}{25} \cdot \frac{5}{4}$$

But then you wrote:
$$\frac{4}{25} \cdot \frac{5}{4} = \frac{16 + 125}{100} = \frac{141}{100} \qquad \leftarrow \textbf{Wrong}$$

The expression $\frac{16+125}{100}$ is the result of **adding** $\frac{4}{25}$ and $\frac{5}{4}$ using a common denominator of 100 — not multiplying them. You wrote a multiplication sign ($\cdot$) but then carried out an addition.

The correct final step:
$$\frac{4}{25} \cdot \frac{5}{4} = \frac{4 \times 5}{25 \times 4} = \frac{20}{100} = \boxed{\frac{1}{5}}$$

**Check:** alternatively, write the tail as $\displaystyle\sum_{n=0}^\infty 4\left(\frac{1}{5}\right)^n - 4 - \frac{4}{5} = \frac{4}{1-1/5} - 4 - \frac{4}{5} = 5 - 4 - \frac{4}{5} = 1 - \frac{4}{5} = \frac{1}{5}$ ✓

This is a common slip under time pressure: writing $\cdot$ and computing $+$. Worth pausing after a fraction multiplication to ask: *is my numerator smaller than either denominator?* Here $20 < 25$, confirming the result is less than 1, which makes sense since the original series terms shrink rapidly.

### Q1(d)

**Key content correct; proof direction slightly circular.**

Your limit argument is exactly right: write
$$S_N = \frac{1}{1-r} - \frac{r^{N+1}}{1-r}$$
and since $|r| < 1$ implies $r^{N+1} \to 0$ as $N \to \infty$, the second term vanishes and $\lim_{N\to\infty} S_N = \frac{1}{1-r}$. That is the core of the argument. ✓

The one structural issue: you started from the formula $\displaystyle\sum_{n=0}^N r^n = \frac{1-r^{N+1}}{1-r}$ and multiplied both sides by $(1-r)$. But that just rearranges the formula you were asked to verify — it doesn't establish it. The question was asking you to *derive* the partial sum formula first, then take the limit.

The clean approach is:

Let $S_N = 1 + r + r^2 + \cdots + r^N$. Then:
$$r S_N = r + r^2 + \cdots + r^N + r^{N+1}$$
Subtract:
$$(1 - r)S_N = S_N - rS_N = 1 - r^{N+1}$$
So for $r \neq 1$:
$$S_N = \frac{1 - r^{N+1}}{1-r}$$

Then the infinite series formula follows exactly as you said: since $|r| < 1$, $r^{N+1} \to 0$, giving
$$\sum_{n=0}^\infty r^n = \lim_{N\to\infty} S_N = \frac{1}{1-r}$$

The multiplication trick (multiply $S_N$ by $r$, then subtract) is the standard derivation and worth knowing as a technique in its own right — telescoping by shifting and subtracting.

### Q1(e)

**Correct, and cleanly executed.**

You recognised that $(-1)^n x^{2n} = (-x^2)^n$, so the series is geometric with ratio $r = -x^2$, and:
$$\sum_{n=0}^\infty (-x^2)^n = \frac{1}{1-(-x^2)} = \frac{1}{1+x^2}$$

The condition $|r| < 1$ becomes $x^2 < 1$, i.e. $|x| < 1$, matching the given domain. ✓

This result — $\displaystyle\frac{1}{1+x^2} = \sum_{n=0}^\infty (-1)^n x^{2n}$ — is important: it is the starting point for deriving the Maclaurin series for $\arctan x$ by integration, which appears in Q4(d).
