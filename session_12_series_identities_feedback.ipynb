Multiplying both sides by $(1-r)$ is exactly what the question asked for, and your limit argument is right. The gap is that you stopped one step short: multiplying both sides of the claimed formula by $(1-r)$ gives

$$(1-r) S_N = 1 - r^{N+1}$$

but you then just divided back by $(1-r)$ to recover the formula. The missing step is expanding the **left-hand side** to show that $(1-r)S_N$ actually equals $1 - r^{N+1}$ — that expansion is the verification:

$$(1-r)\sum_{n=0}^N r^n = \bigl(1 + r + r^2 + \cdots + r^N\bigr) - \bigl(r + r^2 + \cdots + r^N + r^{N+1}\bigr) = 1 - r^{N+1}$$

Every interior term cancels (this is the telescoping); only the first term of the first bracket and the last term of the second bracket survive. That cancellation is what *establishes* the equality — without it, you are asserting rather than proving.

Once you have $(1-r)S_N = 1 - r^{N+1}$, dividing by $(1-r)$ and rearranging gives

$$S_N = \frac{1}{1-r} - \frac{r^{N+1}}{1-r}$$

and then your conclusion is exactly right: since $|r| < 1$, $r^{N+1} \to 0$ as $N \to \infty$, so

$$\sum_{n=0}^\infty r^n = \lim_{N\to\infty} S_N = \frac{1}{1-r}$$

In short: the approach was correct; the telescoping expansion was the one missing line.

### Q2(d)

**Correct.**

$\ln L = \lim_{n\to\infty} \frac{\ln n}{n}$; L'Hôpital gives $\frac{1/n}{1} \to 0$, so $L = e^0 = 1$. ✓

**What $n^{1/n} \to 1$ tells you about polynomials vs exponentials.**

The direct claim is: for any fixed $c > 1$ and any fixed power $k$, the polynomial $n^k$ is eventually dominated by the exponential $c^n$:
$$\frac{n^k}{c^n} \to 0 \quad \text{as } n \to \infty$$
This is why series like $\sum n^3/5^n$ converge — the exponential denominator swamps the polynomial numerator.

The connection to $n^{1/n} \to 1$ is through the **root test**. Apply it to $\sum n^k/c^n$:
$$\left(\frac{n^k}{c^n}\right)^{1/n} = \frac{\left(n^{1/n}\right)^k}{c} \to \frac{1^k}{c} = \frac{1}{c} < 1$$

The polynomial factor $n^k$ contributes $(n^{1/n})^k \to 1^k = 1$ — it vanishes entirely from the root test limit. Only the exponential $c^n$ survives, contributing $1/c$. So $n^{1/n} \to 1$ is precisely what allows you to discard any polynomial factor when an exponential is present. The original comment skipped this connecting step and was therefore unclear.

## Q3 — The constant-inside-a-power error: targeted drill

### Q3(a)

**Correct.**

You correctly factored out $\frac{1}{4}$ before raising the remaining ratio to the power:
$$\frac{(n+1)^3}{4n^3} = \frac{1}{4}\left(\frac{n+1}{n}\right)^3, \qquad c = 4,\ k = 3$$

---

### Q3(b)

**Correct.**

The ratio test calculation:
$$\frac{a_{n+1}}{a_n} = \frac{(n+1)^3}{4^{n+1}} \cdot \frac{4^n}{n^3} = \frac{(n+1)^3}{4n^3} = \frac{1}{4}\left(\frac{n+1}{n}\right)^3 \to \frac{1}{4} < 1$$

The series converges absolutely. ✓ The constant-factoring step is shown explicitly as required.

---

### Q3(c)

**Correct on all three parts.**

**(i)** The student's (wrong) expression gives:
$$\left(\frac{n+1}{7n}\right)^4 = \left(\frac{1}{7} + \frac{1}{7n}\right)^4 \to \left(\frac{1}{7}\right)^4 = \frac{1}{2401}$$

**(iii)** The student concludes convergence (from $\frac{1}{2401} < 1$). The correct analysis also concludes convergence (from $\frac{1}{7} < 1$).

It is worth understanding *why* the error does not change the convergence conclusion here — and in fact, for this type of ratio (polynomial over exponential), it never does. The correct limit is $1/c$ and the student's wrong limit is $1/c^k$. For any $c > 1$: both $1/c < 1$ and $1/c^k < 1$, so both give convergence. For any $c < 1$: both exceed 1, giving divergence. The error consistently produces the wrong numerical value but the same qualitative verdict. The reason it still matters: the wrong limit would be reported as a fact (e.g., in a proof or worked solution), and it could matter in contexts where the exact rate of convergence is needed, not just the yes/no answer.
