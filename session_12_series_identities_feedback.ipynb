### Q4(d) — Elaboration: closed form and Abel's theorem

#### What "closed form" means

A **closed-form expression** is one written using a finite number of standard operations: arithmetic, powers, roots, exponentials, logarithms, trig functions. The key word is *finite* — you can in principle evaluate it in a fixed number of steps, with no infinite process.

| Expression | Closed form? |
|---|---|
| $\dfrac{1}{1+x^2}$ | ✓ — four operations |
| $\arctan(x)$ | ✓ — a named standard function |
| $\displaystyle\sum_{n=0}^\infty (-1)^n x^{2n}$ | ✗ — infinite sum |
| $1 - \tfrac{1}{3} + \tfrac{1}{5} - \cdots$ | ✗ — infinite sum |
| $\dfrac{\pi}{4}$ | ✓ — a single constant |

When you integrate the series for $\dfrac{1}{1+x^2}$, the right side is still an infinite series. The left side, $\arctan(x)$, is a closed form. That is what makes it useful: at $x = 1$, you can evaluate $\arctan(1) = \dfrac{\pi}{4}$ exactly, turning the infinite series on the right into a single number.

The phrase "write a closed form for $\sum_{n=1}^\infty nx^{n-1}$" (Q4(b)) means the same thing: replace the infinite series with a single expression — in that case $\dfrac{1}{(1-x)^2}$ — that you can evaluate directly.

---

#### Why the endpoint $x = 1$ requires a separate argument

The power series for $\arctan$ was derived by integrating $\sum (-1)^n x^{2n}$, which converges only for $|x| < 1$. The equality

$$\arctan(x) = \sum_{n=0}^\infty \frac{(-1)^n x^{2n+1}}{2n+1}$$

is therefore only guaranteed *strictly inside* the interval $(-1, 1)$.

To evaluate at $x = 1$, you need to answer two separate questions:

1. **Does the series $\displaystyle\sum_{n=0}^\infty \frac{(-1)^n}{2n+1}$ converge at all?**  
   Yes — by the Alternating Series Test ($b_n = \frac{1}{2n+1}$ is decreasing and $\to 0$).

2. **Does it converge to $\arctan(1) = \dfrac{\pi}{4}$, or to something else?**  
   This is what Abel's theorem answers.

#### Abel's theorem

> **Abel's theorem:** If a power series $\displaystyle\sum_{n=0}^\infty a_n x^n$ has radius of convergence $R$, and it converges at the endpoint $x = R$, then
> $$\lim_{x \to R^-} \sum_{n=0}^\infty a_n x^n = \sum_{n=0}^\infty a_n R^n$$
> In other words, the power series function is **continuous at the endpoint from the left**.

The statement says: if the series converges at the boundary, you can pass the limit inside the sum. The sum at $x=R$ equals the limit of the closed-form function as $x$ approaches $R$ from inside.

#### Applying this to Q4(d)

- Inside $|x| < 1$: $f(x) = \arctan(x) = \displaystyle\sum_{n=0}^\infty \frac{(-1)^n x^{2n+1}}{2n+1}$.
- At $x = 1$: the series $\displaystyle\sum_{n=0}^\infty \frac{(-1)^n}{2n+1}$ converges (by AST).
- Abel's theorem therefore guarantees: $\displaystyle\sum_{n=0}^\infty \frac{(-1)^n}{2n+1} = \lim_{x \to 1^-} \arctan(x) = \arctan(1) = \frac{\pi}{4}$.

#### Why does this matter — could it go wrong without it?

Yes. A power series can converge at an endpoint to a value that is *not* what the closed-form function gives there, if the closed-form function itself is discontinuous at that point. Abel's theorem rules this out when the function is continuous: since $\arctan$ is continuous everywhere, the series must converge to exactly $\arctan(1)$.

A simpler analogy: if you know that $f(x) = g(x)$ for all $x$ in $(0,1)$, and both $f$ and $g$ are continuous at $x=1$, then $f(1) = g(1)$. Abel's theorem is the version of this continuity argument for power series at their radius of convergence.
