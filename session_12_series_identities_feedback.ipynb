The key theorem this exercise illustrates:

> **Term-by-term differentiation theorem:** If $f(x) = \sum_{n=0}^\infty a_n x^n$ converges for $|x| < R$, then $f$ is differentiable and $f'(x) = \sum_{n=1}^\infty n a_n x^{n-1}$ for $|x| < R$. The radius of convergence of the differentiated series is the same $R$.

The elegant feature of this question is the **roundtrip**: part (c) went from $\frac{1}{1+x}$ to $\ln(1+x)$ by integration; part (e) came back from $\ln(1+x)$ to $\frac{1}{1+x}$ by differentiation. Integration and differentiation are inverses of each other for power series, just as they are for ordinary functions — and both operations preserve the radius of convergence.

### Q4(a) — Elaboration: proving $f'(x) = f(x)$ via index re-labelling

Your discomfort is well-placed. The awkwardness came from choosing the index shift in the wrong direction. Here is the clean proof, followed by an explanation of why direction matters.

---

#### The clean proof

**Step 1 — differentiate term by term:**
$$f(x) = \sum_{n=0}^\infty \frac{x^n}{n!}$$
$$f'(x) = \sum_{n=0}^\infty \frac{d}{dx}\frac{x^n}{n!} = \sum_{n=0}^\infty \frac{n\,x^{n-1}}{n!}$$

The $n=0$ term is $\dfrac{0 \cdot x^{-1}}{0!} = 0$, so drop it:
$$f'(x) = \sum_{n=1}^\infty \frac{n\,x^{n-1}}{n!}$$

Simplify $\dfrac{n}{n!} = \dfrac{1}{(n-1)!}$:
$$f'(x) = \sum_{n=1}^\infty \frac{x^{n-1}}{(n-1)!}$$

**Step 2 — re-label with $m = n - 1$:**

The substitution $m = n-1$ means $n = m+1$. Everything translates mechanically:

| Original (in $n$) | Translated (in $m = n-1$) |
|---|---|
| Lower bound: $n = 1$ | $m = 0$ |
| Upper bound: $n \to \infty$ | $m \to \infty$ |
| Exponent: $x^{n-1}$ | $x^m$ |
| Factorial: $(n-1)!$ | $m!$ |

Substituting:
$$f'(x) = \sum_{m=0}^\infty \frac{x^m}{m!} = f(x) \quad \checkmark$$

---

#### Why this is not a trick — the dummy variable principle

The index $n$ in $\displaystyle\sum_{n=0}^\infty a_n$ is a **dummy variable** — it is internal to the sum and does not appear anywhere outside it. Renaming it changes nothing, just as renaming the variable of integration:
$$\int_0^1 x^2\,dx \;=\; \int_0^1 t^2\,dt$$
These compute the same number. Similarly, $\displaystyle\sum_{n=0}^\infty \frac{x^n}{n!} = \sum_{m=0}^\infty \frac{x^m}{m!}$ — the same series, different label. Re-indexing is renaming, and renaming is always valid.

---

#### Why direction matters: $m = n-1$ versus $m = n+1$

When you differentiate $x^n/n!$ you get $x^{n-1}/(n-1)!$ — the exponent **decreased by 1**. To turn $x^{n-1}$ back into a clean $x^m$, you need $m = n-1$, which shifts the index **down by 1**. The lower bound transforms as $n=1 \mapsto m=0$, landing back at $m=0$ to match $f(x)$ exactly.

If instead you substitute $m = n+1$ (shifting **up**), the translation table becomes:

| Original (in $n$) | Translated (in $m = n+1$) |
|---|---|
| Lower bound: $n = 1$ | $m = 2$ |
| Exponent: $x^{n-1}$ | $x^{m-2}$ |
| Factorial: $(n-1)!$ | $(m-2)!$ |

The lower bound is now $m=2$, not $m=0$, so the resulting series is missing its first two terms relative to $f(x)$. To recover $f(x)$ you have to add those missing terms back by hand, which is the awkwardness you experienced. The algebra still works out, but you are fighting the natural direction of the substitution the whole way.

**Rule of thumb:** after differentiating, the exponent decreases by 1, so substitute $m = n-1$ (shift down) to re-align it. After integrating, the exponent increases by 1, so substitute $m = n+1$ (shift up) to re-align it.

#### What "consistent with the sum from $n=0$ to $\infty$" means

The concern is: after the re-indexing, do we still have an infinite series of the same type? Yes, and here is why.

The original $f(x)$ runs from $n=0$ to $\infty$, producing terms $1,\ x,\ x^2/2!,\ x^3/3!,\ \ldots$ After differentiation and re-indexing with $m = n-1$, $f'(x)$ also runs from $m=0$ to $\infty$, producing $1,\ x,\ x^2/2!,\ x^3/3!,\ \ldots$ — the identical terms in the identical order. Re-indexing is a one-to-one relabelling of the sequence; it does not rearrange, skip, or duplicate any terms. Two series are equal when their terms match position by position, and that is exactly what holds here.

### Q4(d) — Elaboration: closed form, what "converges" means, and Abel's theorem

#### What "closed form" means

A **closed-form expression** is one written using a finite number of standard operations: arithmetic, powers, roots, exponentials, logarithms, trig functions. The key word is *finite* — you can evaluate it in a fixed number of steps, with no infinite process.

| Expression | Closed form? |
|---|---|
| $\dfrac{1}{1+x^2}$ | ✓ — four operations |
| $\arctan(x)$ | ✓ — a named standard function |
| $\displaystyle\sum_{n=0}^\infty (-1)^n x^{2n}$ | ✗ — infinite sum |
| $1 - \tfrac{1}{3} + \tfrac{1}{5} - \cdots$ | ✗ — infinite sum |
| $\dfrac{\pi}{4}$ | ✓ — a single constant |

When you integrate the series for $\dfrac{1}{1+x^2}$, the right side is still an infinite series. The left side, $\arctan(x)$, is a closed form. That is what makes it useful: at $x = 1$ you can evaluate $\arctan(1) = \dfrac{\pi}{4}$ exactly, turning the infinite series into a single number.

---

#### What "the series converges" means — partial sums

"The series $\sum_{n=0}^\infty \frac{(-1)^n}{2n+1}$ converges" does **not** mean a partial sum up to some specific point equals a certain value. It means something about what happens as you keep adding *infinitely many* terms.

Define the partial sums:
$$S_0 = 1, \qquad S_1 = 1 - \tfrac{1}{3}, \qquad S_2 = 1 - \tfrac{1}{3} + \tfrac{1}{5}, \qquad S_3 = 1 - \tfrac{1}{3} + \tfrac{1}{5} - \tfrac{1}{7}, \quad \ldots$$

Each $S_N$ is a finite sum with a definite numerical value — you can compute it exactly. But none of them *is* the infinite series; each is only an approximation. **The series converges** means: as $N \to \infty$, the sequence $S_0, S_1, S_2, \ldots$ approaches a specific limiting value $L$, and stays arbitrarily close to it. That limit $L$ is what we call the sum of the series.

The phrase "the series converges *at* $x = 1$" just means: substitute $x = 1$ into the power series to get a specific sequence of numbers to sum, then ask whether that sequence of partial sums has a limit. It tells you *which* series you are asking about, not that you stop at a particular term.

---

#### Why the endpoint $x = 1$ requires a separate argument

The series $\arctan(x) = \displaystyle\sum_{n=0}^\infty \frac{(-1)^n x^{2n+1}}{2n+1}$ was derived from a geometric series that converges only for $|x| < 1$, so this equality is only guaranteed *strictly inside* the interval. At $x = 1$ exactly, two separate questions arise:

1. **Does the series $\displaystyle\sum_{n=0}^\infty \frac{(-1)^n}{2n+1}$ converge?**  
   Substitute $x=1$, form the partial sums, check whether they approach a limit. Yes — by the AST the partial sums converge.

2. **Does the series converge to $\arctan(1) = \pi/4$, or could it converge to something else?**  
   This is the non-trivial question, and it is what Abel's theorem answers.

#### Abel's theorem

> If a power series $\displaystyle\sum_{n=0}^\infty a_n x^n$ has radius of convergence $R$, and the series of numbers $\displaystyle\sum_{n=0}^\infty a_n R^n$ converges (i.e. the partial sums at $x = R$ have a limit), then that limit equals $\displaystyle\lim_{x \to R^-} f(x)$ where $f(x)$ is the closed-form function the series represents inside $|x| < R$.

In plain terms: if the partial sums converge at the boundary, they converge to the value you get by approaching the boundary from inside.

#### Applying this to Q4(d)

- For $|x| < 1$: $\arctan(x) = \displaystyle\sum_{n=0}^\infty \frac{(-1)^n x^{2n+1}}{2n+1}$ (established by integration).
- At $x = 1$: the partial sums of $\displaystyle\sum_{n=0}^\infty \frac{(-1)^n}{2n+1}$ converge (verified by AST).
- Abel's theorem: those partial sums converge to $\lim_{x \to 1^-} \arctan(x) = \arctan(1) = \dfrac{\pi}{4}$.

So $\dfrac{\pi}{4} = 1 - \dfrac{1}{3} + \dfrac{1}{5} - \cdots$ is not just a notational convenience — it is backed by this continuity argument.

### Q6(d) — Elaboration: how to choose the comparison series in the LCT

#### Your insight

You noted that "drop minor terms" is a heuristic for finding the comparison series, and asked whether you are free to choose any series you like provided you know its convergence behaviour. That is essentially right — the comparison series is a strategic choice, not a forced one. Here is the precise version.

---

#### The two requirements for a valid LCT comparison

The Limit Comparison Test says: if $a_n > 0$, $b_n > 0$, and

$$L = \lim_{n\to\infty} \frac{a_n}{b_n}, \qquad 0 < L < \infty$$

then $\sum a_n$ and $\sum b_n$ either **both converge or both diverge**.

For the test to be useful, your choice of $b_n$ must satisfy **both** of the following:

1. **You know whether $\sum b_n$ converges or diverges** — otherwise the conclusion gives you nothing.
2. **The ratio $a_n/b_n$ has a finite positive limit** — if the limit is $0$, $\infty$, or does not exist, the test does not apply.

You are free to choose any $b_n$ satisfying both. There is no uniquely correct choice.

---

#### "Drop minor terms" is a systematic way to satisfy requirement 2

For $a_n = \dfrac{n}{n^2+1}$, the dominant term in the numerator is $n$ and in the denominator is $n^2$, so

$$\frac{n}{n^2+1} \;\sim\; \frac{n}{n^2} = \frac{1}{n} \qquad \text{as } n \to \infty$$

Choosing $b_n = 1/n$ then gives $a_n/b_n \to 1 \in (0,\infty)$ — requirement 2 is satisfied automatically, by construction. Requirement 1 is satisfied because $\sum 1/n$ is the harmonic series, which you know diverges.

You could equally choose $b_n = 2/n$, $b_n = 1/(n+1)$, $b_n = 5/n$ — any series with the same asymptotic scale:

$$\frac{n/(n^2+1)}{c/n} = \frac{n^2}{c(n^2+1)} \to \frac{1}{c} \in (0,\infty) \quad \text{for any constant } c > 0$$

All of these are valid comparisons. "Drop minor terms" just lands you on the simplest choice ($c = 1$) automatically.

---

#### The heuristic can mislead near the boundary of convergence

"Drop minor terms" gives you a candidate; it does not guarantee the LCT will apply. It can mislead when the dominant term sits exactly on the p-series boundary ($p = 1$).

**Example:** $\displaystyle\sum_{n=2}^\infty \frac{1}{n \ln n}$

Dropping minor terms gives $a_n \sim 1/n$, so you might try $b_n = 1/n$:
$$\frac{1/(n \ln n)}{1/n} = \frac{1}{\ln n} \to 0$$

The limit is $0$, not a positive finite number — requirement 2 fails and the LCT does not apply. The series $\sum 1/(n \ln n)$ is *smaller* than the harmonic series, so a comparison with the harmonic series only tells you it *might* converge — it cannot confirm divergence. (It actually diverges, but you need the integral test to prove it.)

The moral: "drop minor terms" gives you a candidate; you still need to verify the ratio limit is positive and finite before the LCT applies.

---

#### Summary

| Requirement | What satisfies it |
|---|---|
| $\sum b_n$ has known behaviour | Choose a standard series: p-series, harmonic, geometric |
| $a_n/b_n \to L \in (0,\infty)$ | Choose $b_n$ with the same dominant asymptotic scale as $a_n$ ("drop minor terms") |

Your insight is correct: the comparison series is a *strategic choice*. Any $b_n$ satisfying both requirements is valid. "Drop minor terms" is a principled strategy for finding the simplest such choice, not a rule that produces the unique answer.

### Q6(f) — Elaboration: definition vs consequence in absolute and conditional convergence

#### The issue with defining conditional convergence by rearrangement

A natural but slightly off framing is to say "conditional convergence means that rearranging the terms can change the sum." That describes the *consequence* of conditional convergence, not its *definition*. The distinction matters because you cannot check the rearrangement property directly in practice — you check the definition, and the rearrangement property follows as a theorem.

---

#### The definitions (modern standard)

**Plain convergence:** $\sum a_n$ converges if its partial sums approach a finite limit:
$$S_N = \sum_{n=1}^N a_n \;\longrightarrow\; S \in \mathbb{R} \quad \text{as } N \to \infty$$
This is the base concept — absolute and conditional convergence are both refinements of it.

**Absolute convergence:** $\sum a_n$ is absolutely convergent if $\sum |a_n|$ converges.

**Theorem (absolute ⟹ plain):** If $\sum a_n$ is absolutely convergent, then it is convergent.

**Conditional convergence:** $\sum a_n$ is conditionally convergent if:
- $\sum a_n$ converges (plain), **and**
- $\sum |a_n|$ diverges.

That is the complete definition — it makes no reference to rearrangements.

**Riemann rearrangement theorem:** *If* $\sum a_n$ is conditionally convergent (by the above definition), *then* for any $L \in \mathbb{R}$ there exists a rearrangement of the terms that converges to $L$, and there also exist rearrangements that diverge to $+\infty$ or $-\infty$.

The rearrangement property is what *follows from* being conditionally convergent. It is not what conditional convergence means.

---

#### The full hierarchy

```
All series
├── Divergent        partial sums have no finite limit
└── Convergent       partial sums → finite limit S
    ├── Absolutely convergent     ∑|aₙ| converges
    │     └── any rearrangement also converges to S  (Dirichlet's theorem)
    └── Conditionally convergent  ∑|aₙ| diverges
          └── rearrangements can sum to any value, or diverge  (Riemann's theorem)
```

Under the modern standard, absolutely convergent and conditionally convergent are **mutually exclusive** subcategories of convergent — a series cannot be both.

---

#### How to classify a series in practice

1. Test $\sum |a_n|$ first.
   - If $\sum |a_n|$ converges → **absolutely convergent**; done.
   - If $\sum |a_n|$ diverges → move to step 2.
2. Test $\sum a_n$ directly (AST, Dirichlet's test, etc.).
   - If $\sum a_n$ converges → **conditionally convergent**.
   - If $\sum a_n$ also diverges → **divergent**.

The rearrangement theorem is a property you can then *quote* for conditionally convergent series — not something you use to establish the classification.

Whether the original series $\sum \cos(n\theta)/n$ converges at all (without absolute values) is a separate question that requires a different test — Dirichlet's test — which is beyond the scope of this session.

### Primer: Recognising series and finding their closed forms

This primer gives you the vocabulary and toolkit you need before attempting Q7(e).

---

#### Part 1 — The core series lexicon

These are the series worth knowing by heart. Everything in Q7(e) — and most of Fourier analysis — is built from them.

| Series | Closed form | Notes |
|---|---|---|
| $\displaystyle\sum_{n=0}^\infty \frac{c^n}{n!}$ | $e^c$ | Any real $c$; factorial denominator is the giveaway |
| $\displaystyle\sum_{n=0}^\infty r^n$ | $\dfrac{1}{1-r}$ | Geometric; requires $|r| < 1$ |
| $\displaystyle\sum_{n=0}^\infty (-1)^n x^n$ | $\dfrac{1}{1+x}$ | Geometric with $r = -x$; requires $|x|<1$ |
| $\displaystyle\sum_{n=0}^\infty (-1)^n x^{2n}$ | $\dfrac{1}{1+x^2}$ | Geometric with $r = -x^2$ |
| $\displaystyle\sum_{n=1}^\infty \frac{(-1)^{n+1}}{n} x^n$ | $\ln(1+x)$ | Integration of the geometric series for $1/(1+x)$ |
| $\displaystyle\sum_{n=0}^\infty \frac{(-1)^n x^{2n+1}}{2n+1}$ | $\arctan(x)$ | Integration of the series for $1/(1+x^2)$ |

**Special evaluations worth knowing** (derived by substituting a specific value of $x$ or $c$):

$$e^{-1} = \sum_{n=0}^\infty \frac{(-1)^n}{n!}, \qquad \ln 2 = \sum_{n=1}^\infty \frac{(-1)^{n+1}}{n}, \qquad \frac{\pi}{4} = \sum_{n=0}^\infty \frac{(-1)^n}{2n+1}$$

---

#### Part 2 — A recognition decision tree

When you see an unfamiliar series, work through these questions:

**Question 1: Is there a factorial $n!$ in the denominator?**
- Yes → almost certainly the exponential series $\sum c^n/n!$. The numerator tells you $c$.
- No → continue.

**Question 2: What is the sign pattern?**
- All positive terms → geometric $\sum r^n$ (positive $r$), or exponential $\sum c^n/n!$ with $c > 0$.
- Alternating $(-1)^n$ → something involving a negative substitution.

**Question 3: What is the denominator?**
- Just $r^n$ (no $n$ in denominator) → geometric; identify $r$.
- Just $n$ → likely from $\ln(1+x)$ evaluated at some $x$.
- Odd numbers $2n+1$ → almost certainly from $\arctan$.
- $n!$ (factorial) → exponential, as above.

**Question 4: Is there a constant factor or a shifted starting index?**
- Constant $\times$ series → factor it out first, identify the standard part, then multiply the closed form.
- Sum starts at $n = k$ instead of $n = 0$ → either (a) the $n=0$ term happened to be zero (safe to restore), or (b) you are missing the first $k$ terms: $\sum_{n=k}^\infty a_n = \sum_{n=0}^\infty a_n - a_0 - \cdots - a_{k-1}$.

---

#### Part 3 — The key manipulation: substituting into a known series

The most powerful technique is **substitution**: take a known series identity and replace the variable with something specific.

**Example A** — identifying via substitution into the exponential series:

$$\sum_{n=0}^\infty \frac{x^n}{n!} = e^x \quad \text{for all } x$$

To evaluate $\displaystyle\sum_{n=0}^\infty \frac{(-2)^n}{n!}$: compare with $\sum c^n/n!$. The numerator is $(-2)^n$, so $c = -2$. Answer: $e^{-2}$.

**Example B** — identifying via substitution into the geometric series:

$$\sum_{n=0}^\infty r^n = \frac{1}{1-r} \quad \text{for } |r| < 1$$

To evaluate $\displaystyle\sum_{n=0}^\infty \left(\frac{1}{4}\right)^n$: this is geometric with $r = \frac{1}{4}$. Answer: $\dfrac{1}{1-\frac{1}{4}} = \dfrac{4}{3}$.

If there is a constant factor $\lambda$ in every term: $\displaystyle\sum_{n=0}^\infty \lambda \cdot r^n = \lambda \cdot \frac{1}{1-r}$.

**Example C** — identifying via a known special value:

$$\ln(1+x) = \sum_{n=1}^\infty \frac{(-1)^{n+1}}{n} x^n$$

To evaluate $\displaystyle\sum_{n=1}^\infty \frac{(-1)^{n+1}}{n} \cdot \left(\frac{1}{2}\right)^n$: compare with $\ln(1+x)$, substituting $x = \frac{1}{2}$. Answer: $\ln\!\left(1 + \frac{1}{2}\right) = \ln\frac{3}{2}$.

---

#### Part 4 — Checking your answer

After identifying a closed form, verify it in two ways:

1. **Expand a few terms of the series and confirm they match the Taylor expansion of the closed form at $x=0$.** For example, $e^{-1} \approx 1 - 1 + \frac{1}{2} - \frac{1}{6} + \cdots \approx 0.368$; the series $\sum (-1)^n/n!$ has partial sums $1, 0, 0.5, 0.333, 0.375, \ldots$ converging toward $0.368$.

2. **Check dimensions and signs are consistent.** A series with all positive terms cannot converge to a negative number. A series whose terms involve $1/n!$ cannot converge to a simple rational number like $\frac{1}{2}$ or $\frac{1}{3}$.

---

#### Part 5 — What to do when nothing matches

If none of the standard forms seem to fit:

1. **Rewrite a few terms explicitly** to see the pattern more clearly.
2. **Factor out anything that doesn't depend on $n$** (constants, fixed fractions).
3. **Ask whether the numerator looks like it was produced by differentiating or integrating a simpler numerator.** A numerator of $n$ suggests differentiation; a numerator of $1/(n+1)$ or $1/n$ suggests integration.
4. **Check whether the series could be a standard one evaluated at a specific point** (e.g., $x = 1$, $x = -1$, $x = 2$).

With Q7(e), all five series in the table can be matched to one of the closed forms listed in the options using the tools in Parts 1–3 above.

### Elaboration: geometric series vs. Maclaurin series — how to decide

The primer gives you the vocabulary. This note sharpens the specific decision you asked about.

---

#### The single sharpest question

Compute (or eyeball) the ratio of consecutive terms:

$$\rho_n = \frac{a_{n+1}}{a_n}$$

| $\rho_n$ behaviour | What it tells you |
|---|---|
| Constant — no $n$ at all | **Geometric series.** Use $\sum r^n = 1/(1-r)$ directly. |
| Depends on $n$ | **Maclaurin series** of some function. The shape of the $n$-dependence tells you which one. |

This is the ratio test repurposed as a *recognition* tool, not a convergence tool.

---

#### Geometric series in depth

A series is geometric when each term is a fixed multiple of the one before it:

$$a_{n+1} = r \cdot a_n \quad \Longleftrightarrow \quad a_n = a_0 \cdot r^n$$

The closed form follows immediately from the finite geometric sum:

$$\sum_{n=0}^\infty a_0 r^n = \frac{a_0}{1-r}, \qquad |r| < 1$$

**Diagnostic signals:**
- No $n$ in the denominator (unless it cancels out).
- Terms look like $c^n$, $(\text{fraction})^n$, or $A^n B^n$.
- Consecutive ratio $\rho_n$ collapses to a pure number.

In Q7(e), series 5 — $\sum \left(\tfrac{1}{3}\right)^n \cdot \tfrac{1}{2}$ — is geometric with $r = \tfrac{1}{3}$, $a_0 = \tfrac{1}{2}$:

$$\rho_n = \frac{(1/3)^{n+1} \cdot (1/2)}{(1/3)^n \cdot (1/2)} = \frac{1}{3} \quad \text{(constant)}$$

---

#### Maclaurin series in depth

A Maclaurin series is a power series $\sum a_n$ where the ratio $\rho_n = a_{n+1}/a_n$ depends on $n$. The shape of that dependence identifies the function.

| Consecutive ratio $\rho_n$ | Closed form | Example series |
|---|---|---|
| $\dfrac{c}{n+1}$ | $e^c$ | $\sum \dfrac{c^n}{n!}$; ratio $= c/(n+1)$ |
| $-\dfrac{x}{n+1} \cdot \dfrac{n}{1}$ (alternating, $\sim 1/n$) | $\ln(1+x)$ | $\sum \dfrac{(-1)^{n+1}}{n}$; ratio $= -n/(n+1)$ |
| $-\dfrac{x^2(2n+1)}{2n+3}$ (alternating, $\sim 1/n$) | $\arctan(x)$ | $\sum \dfrac{(-1)^n}{2n+1}$; ratio $= -(2n+1)/(2n+3)$ |

In practice you rarely need to compute $\rho_n$ for Maclaurin series — the structural fingerprints are enough:

| Structural fingerprint | Likely match |
|---|---|
| $n!$ in denominator | Exponential $e^c$ |
| Alternating, plain $n$ in denominator | $\ln(1+x)$ |
| Alternating, odd integers $2n+1$ in denominator | $\arctan(x)$ |
| Alternating, $(2n)!$ or $(2n+1)!$ in denominator | $\cos x$ or $\sin x$ |

---

#### When both seem to apply

Geometric IS a Maclaurin series — specifically, it is the Maclaurin series of $1/(1-r)$ evaluated at the given $r$. So $\sum (1/3)^n$ could be described either as geometric with ratio $1/3$, or as $\frac{1}{1-x}$ evaluated at $x = 1/3$. They are the same thing.

**Rule**: when the series fits the geometric template directly (constant ratio, no $n$ in denominator), use the geometric formula. It is simpler and requires no knowledge of which function it corresponds to. Reaching for the Maclaurin framing only adds steps.

The Maclaurin framing is essential only when the series *cannot* be expressed as $\sum r^n$ with a constant $r$ — i.e., when the ratio between terms depends on $n$.

---

#### Applied to Q7(e) — why each tool was the right one

| Series | Consecutive ratio | Tool used | Why |
|---|---|---|---|
| $\sum (-1)^n/n!$ | $(-1)/(n+1)$ — depends on $n$ | Maclaurin ($e^c$) | Constant ratio impossible; $n!$ fingerprint |
| $\sum 2^n/n!$ | $2/(n+1)$ — depends on $n$ | Maclaurin ($e^c$) | Same; $c=2$ |
| $\sum (-1)^{n+1}/n$ | $-n/(n+1)$ — depends on $n$ | Maclaurin ($\ln$) | Alternating, plain $n$ denominator |

| $\sum (1/3)^n \cdot (1/2)$ | $1/3$ — constant | Geometric | Constant ratio; no $n$ dependence |

## Elaboration: Index shifting in series — principles, methods, and pitfalls

Re-indexing (also called index shifting) is the act of substituting a new summation variable so that a series starts at a different index or takes a more convenient form. It is a purely algebraic manipulation — the series itself does not change.

---

### Why re-index?

Three main reasons:

1. **To match a standard form.** Closed-form identities are stated with specific starting indices (usually $n=0$ or $n=1$). If your series starts elsewhere, re-indexing lets you read off the closed form directly.

2. **To combine two series term-by-term.** Two series can only be added term-by-term if they share the same index variable and the same starting index.

3. **To simplify the general term.** Sometimes writing $a_{n+k}$ as the new $a_m$ (where $m = n+k$) makes the term easier to work with.

---

### The two basic moves

#### Move 1 — Shift the index variable (rename)

Substitute $m = n + k$ (equivalently $n = m - k$). Every occurrence of $n$ in the summand is replaced by $m - k$, and the limits adjust:

$$\sum_{n=0}^\infty f(n) \xrightarrow{\,m = n+k\,} \sum_{m=k}^\infty f(m-k)$$

Or in the other direction, shift down: substitute $m = n - k$ ($n = m + k$):

$$\sum_{n=k}^\infty f(n) \xrightarrow{\,m = n-k\,} \sum_{m=0}^\infty f(m+k)$$

The dummy variable name ($n$, $m$, $j$, $k$ — whatever) is irrelevant to the value of the sum. After re-indexing you may rename $m \to n$ again if you prefer.

**Example.** Express $\displaystyle\sum_{n=3}^\infty \frac{x^n}{n!}$ in terms of $e^x$.

We know $e^x = \sum_{n=0}^\infty \frac{x^n}{n!}$. The given sum is missing the $n=0,1,2$ terms, so:

$$\sum_{n=3}^\infty \frac{x^n}{n!} = e^x - \frac{x^0}{0!} - \frac{x^1}{1!} - \frac{x^2}{2!} = e^x - 1 - x - \frac{x^2}{2}$$

No variable substitution was needed here — this is the peel-off technique (see below).

---

#### Move 2 — Peel off (or restore) finitely many terms

If a series differs from a standard form only in its starting index, the relationship is:

$$\sum_{n=0}^\infty a_n = a_0 + a_1 + \cdots + a_{k-1} + \sum_{n=k}^\infty a_n$$

Rearranging:

$$\sum_{n=k}^\infty a_n = \sum_{n=0}^\infty a_n - a_0 - a_1 - \cdots - a_{k-1}$$

This is probably the most-used re-indexing move in practice.

**Example.** Evaluate $\displaystyle\sum_{n=2}^\infty \left(\frac{1}{2}\right)^n$.

The full geometric series gives $\sum_{n=0}^\infty (1/2)^n = 1/(1-1/2) = 2$. We peel off the $n=0$ and $n=1$ terms:

$$\sum_{n=2}^\infty \left(\frac{1}{2}\right)^n = 2 - 1 - \frac{1}{2} = \frac{1}{2}$$

---

### A worked re-indexing example (index variable substitution)

**Problem.** Show that $\displaystyle\sum_{n=1}^\infty \frac{x^{n-1}}{(n-1)!} = e^x$.

The summand has $n-1$ appearing twice. Substitute $m = n - 1$, so $n = m + 1$. When $n=1$, $m=0$; as $n \to \infty$, $m \to \infty$:

$$\sum_{n=1}^\infty \frac{x^{n-1}}{(n-1)!} = \sum_{m=0}^\infty \frac{x^m}{m!} = e^x \quad \checkmark$$

The substitution simply renamed every $(n-1)$ as $m$ and shifted the lower limit from 1 to 0.

---

### Combining two series with different starting indices

**Problem.** Simplify $\displaystyle S = \sum_{n=0}^\infty a_{n+1} - \sum_{n=1}^\infty a_n$.

Re-index the first sum: let $m = n+1$, so $n = m-1$. When $n=0$, $m=1$:

$$\sum_{n=0}^\infty a_{n+1} = \sum_{m=1}^\infty a_m$$

Now both sums run from 1 to $\infty$ over $a_m$, so $S = 0$. (This is the telescoping identity in disguise.)

---

### Common errors and how to avoid them

#### Error 1 — Using $n$ as both the summation variable and a fixed label

This was the error in Q7(d). Writing:
$$\sum_{n=0}^\infty a_{n+k} = S - a_n - a_{n+1} - \cdots - a_{n+k-1} \quad \leftarrow \textbf{WRONG}$$
uses $n$ as both the index running over the sum *and* as a fixed subscript in the right-hand side. These are two different things. The correct statement is:
$$\sum_{n=k}^\infty a_n = S - a_0 - a_1 - \cdots - a_{k-1}$$
where $a_0, a_1, \ldots, a_{k-1}$ are fixed numbers (the first $k$ terms of the original series), not expressions involving the summation variable $n$.

**Rule:** once you have written the $\Sigma$ symbol with an index variable, that variable must not appear as a fixed subscript anywhere outside the summand.

---

#### Error 2 — Shifting the limits without shifting the summand (or vice versa)

If you substitute $m = n + k$, both things must change together:

| Component | Before | After ($m = n+k$) |
|---|---|---|
| Summation variable | $n$ | $m$ |
| Lower limit | $0$ | $k$ |
| Every $n$ in summand | $n$ | $m - k$ |

Forgetting to adjust the summand while changing the limits (or vice versa) produces a different series entirely.

---

#### Error 3 — Off-by-one in the peel-off count

This was the error in Q7(d)(i). When a finite sum $\sum_{n=1}^N$ telescopes, the last term remaining is $a_N - a_{N+1}$, not $a_N - a_N$. The closed form is:

$$\sum_{n=1}^N \left(\frac{1}{n} - \frac{1}{n+1}\right) = 1 - \frac{1}{N+1}$$

The $+1$ in the denominator is easy to drop under pressure. A quick check: at $N=1$ the sum has only one term, $1 - \frac{1}{2} = \frac{1}{2}$. The formula gives $1 - \frac{1}{2} = \frac{1}{2}$ ✓. If you wrote $1 - \frac{1}{N}$ you would get $0$ when $N=1$, which is obviously wrong.

---

#### Error 4 — Assuming $\sum_{n=0}^\infty a_{n+k} = \sum_{n=0}^\infty a_n$

Shifting the argument of $a$ does **not** give the same sum. $\sum a_{n+k}$ is the original series with its first $k$ terms removed:

$$\sum_{n=0}^\infty a_{n+k} = \sum_{n=k}^\infty a_n = S - a_0 - \cdots - a_{k-1}$$

This matters whenever $a_0, \ldots, a_{k-1}$ are non-zero.

---

### Quick reference

| Goal | Move | Key step |
|---|---|---|
| Start series at $n=0$ instead of $n=k$ | Peel off or substitute $m = n-k$ | Lower limit and summand both adjust |
| Start series at $n=k$ instead of $n=0$ | Substitute $m = n+k$ | Lower limit $0 \to k$; every $n$ in summand $\to m-k$ |
| Remove leading terms | Peel off: $\sum_{n=k}^\infty a_n = S - a_0 - \cdots - a_{k-1}$ | Identify $S$ as the known full sum |
| Match $a_{n+k}$ pattern | Substitute $m = n+k$ so summand becomes $a_m$; lower limit shifts to $k$ | Rename $m \to n$ afterwards if desired |
| Combine two series | Re-index one so both have identical limits, then add term-by-term | |

The dummy variable name is irrelevant. What matters is that the lower limit and every occurrence of the index in the summand are updated consistently.
