## Q3 — L'Hôpital's rule

L'Hôpital's rule: if $\lim_{x\to a} f(x) = \lim_{x\to a} g(x) = 0$ (or both $\pm\infty$), then
$$\lim_{x\to a}\frac{f(x)}{g(x)} = \lim_{x\to a}\frac{f'(x)}{g'(x)}$$
provided the right-hand limit exists. For forms other than $0/0$ or $\infty/\infty$, rewrite algebraically first.

## Q4 — Volumes of revolution: disk and washer methods

Revolving a region around the $x$-axis generates a solid. The volume is computed by integrating the cross-sectional area:
- **Disk:** $V = \pi\displaystyle\int_a^b [R(x)]^2\,dx$ (one curve, one radius)
- **Washer:** $V = \pi\displaystyle\int_a^b \left([R(x)]^2 - [r(x)]^2\right)\,dx$ (two curves; $R$ outer, $r$ inner)

**(a)** The region bounded by $y = \sqrt{x}$, $x = 0$, $x = 4$, $y = 0$ is revolved around the $x$-axis. Set up and evaluate the volume using the disk method.

**(b)** The region bounded by $y = x$ and $y = x^2$, $0 \leq x \leq 1$, is revolved around the $x$-axis. Identify the outer and inner radii, then set up and evaluate the volume using the washer method.
