# Session 11 — Stokes' Theorem and the Divergence Theorem

**Date:** *(fill in when you work this session)*

Focus: Stokes' Theorem (generalises Green's Theorem to surfaces in 3D — relates a surface integral of curl $\mathbf{F}$ to a line integral around its boundary) and the Divergence Theorem (relates a surface integral of flux to a volume integral of divergence). The key skill is choosing the right theorem and setting up the integrals correctly.

## Q1 — Curl and Stokes' Theorem

Let $\mathbf{F}(x,y,z) = (-y, x, z^2)$.

(a) Compute $\nabla \times \mathbf{F}$ (the curl).

(b) Use Stokes' Theorem to evaluate $\oint_C \mathbf{F} \cdot d\mathbf{r}$ where $C$ is the circle $x^2 + y^2 = 1$ in the plane $z = 1$, traversed counterclockwise when viewed from above.

(Take the surface $S$ to be the flat disk bounded by $C$.)

I start by calculating the Curl of the vector field by taking the cross product of the nabla operator with the vector field

\begin{gather*}
& \nabla \times F=\left|\begin{array}{ccc} \\
\hat{i} & \hat{j} & \hat{k} \\
\frac{\partial}{\partial x} & \frac{\partial}{\partial y} & \frac{\partial}{\partial z} \\
-y & x & z^2
\end{array}\right| \\
& \left(\frac{\partial z^2}{\partial y}-\frac{\partial x}{\partial z}\right) \hat{i}-\left(\frac{\partial z^2}{\partial x}-\frac{\partial-y}{\partial z}\right) \hat{j}+\left(\frac{\partial x}{\partial x}-\frac{\partial-y}{\partial y}\right) \hat{k} \\
& (0-0) \hat{i}-(0-0) \hat{j}+(1-(-1)) \hat{k} \\
& =2 \hat{k} \\
\end{gather*}

I can then drop this value for Curl into Stoekes Theorem and turn the line integral into a two dimensinal integral over a region.

\begin{gather*}
& \oint_C F \cdot d r=\iint_R 2 d A \\
\end{gather*}

As the region is a circular region it is convenient to convert to polar coordinates and change the differential and bounds accordingly

\begin{gather*}
& x^2+y^2=r^2 \\
& r^2=1 \rightarrow r=1 \\
& x=\cos \theta \\
& y=\sin \theta \\
& \int_{\theta=0}^{\theta=2\pi} \int_{r=0}^{r=1} 2r d r d \theta \\
& \int_{\theta=0}^{\theta=2\pi} \left[r^2\right]_{r=0}^{r=1} d \theta \rightarrow \int_{\theta=0}^{\theta=2\pi} 1 d \theta \\
& \left[ \theta \right]_{\theta=0}^{\theta=2\pi} \rightarrow 2\pi \\
\end{gather*}

## Q2 — Choosing the surface

For the same field $\mathbf{F} = (-y, x, z^2)$ and the same curve $C$ as Q1, apply Stokes' Theorem using a different surface — the hemisphere $x^2 + y^2 + z^2 = 1$, $z \geq 0$, bounded by $C$. Show you get the same answer, and explain why this must always be the case.

**Your answer:**
I start by finding the Curl of the vector field by using nabla with the cross product:
\begin{gather*}
& \nabla \times f=0 \hat{i}+0 \hat{j}+2 \hat{k} \\
\end{gather*}
I then solve the surface equation for z giving me a definition of g(x,y) that I take partial derivatives of with respect to x and then y
\begin{gather*}
& z^2=1-x^2-y^2 \\
& z=\sqrt{1-x^2-y^2} \\
& z=g(x, y)=\sqrt{1-x^2-y^2} \\
& g_x=-\frac{x}{\sqrt{1-x^2-y^2}} \\
& g_y=-\frac{y}{\sqrt{1-x^2-y^2}} \\
& \left(-y\left(-\frac{x}{\sqrt{1-x^2-y^2}}\right)\right)=\frac{x y}{\sqrt{1-x^2-y^2}} \\
& \left(x\left(-\frac{y}{\sqrt{1-x^2-y^2}}\right)\right)=-\frac{x y}{\sqrt{1-x^2-y^2}} \\
\end{gather*}

I then compose these results into the Stoke Thorem formula (which is just a formulation of the cross product of the parallelograms on the surface giving surface area of each paralellogram that can be used as a scaling factor from the parameter space to the surface space. In addition it composes the dot product with the norm of each parallelogram so that only the "force" in the same direction as the norm of each parallelogram is included) Formula is -Pg_x - Qg_y + R where P,Q,R come from the Vector field cross product worked out earlier 
\begin{gather*}
& -\left(\frac{x y}{\sqrt{1-x^2-y^2}}\right)(0)-\left(-\frac{x y}{\sqrt{1-x^2-y^2}}\right)(0)+2=2 \\
&\oint_C F \cdot d r=\iint_R 2 d A\\
\end{gather*}
This gives us exactly the same result as with the previous question and we can plug it into the same double intwegral as the Surface is bounded on the xy plane by the same circular boundary and this gives us the same result:
\begin{gather*}
& x^2+y^2=r^2 \\
& r^2=1 \rightarrow r=1 \\
& x=r \cos \theta \\
& y=r \sin \theta \\
&\int_{\theta=0}^{\theta=2 \pi} \int_{r=0}^{r=1} 2 r d r d \theta \\
& \int_{\theta=0}^{\theta=2 \pi}\left[r^2\right]_{r=0}^{r=1} d \theta \rightarrow \int_{\theta=0}^{\theta=2 \pi} 1 d \theta \\
& {[\theta]_{\theta=0}^{\theta=2 \pi} \rightarrow 2 \pi}
\end{gather*}

The results should be the same as Green/Stoke rely on the principle that the curls of adjacent paralellograms cancel each other out and it is only the sides of the parallelograms on the outside edge that arent cancelled out. Therefore the sum of the Curl with the region is mathematically equivalent to the curl from walking the boundary.

## Q3 — Divergence Theorem

Let $\mathbf{F}(x,y,z) = (x^3, y^3, z^3)$ and let $V$ be the ball $x^2 + y^2 + z^2 \leq 1$.

(a) Compute $\nabla \cdot \mathbf{F}$ (the divergence).

(b) Use the Divergence Theorem to evaluate $\iint_S \mathbf{F} \cdot d\mathbf{S}$ where $S$ is the surface of the ball.

(You will need the volume of the ball and the formula $\iiint_V (x^2 + y^2 + z^2) \, dV$ in spherical coordinates.)

**Your answer:**

First I find divergence of the vector field by taking the dot product of the vector field with the nabla operator

\begin{gather*}
& \left(\frac{\partial}{\partial x}, \frac{\partial}{\partial y}, \frac{\partial}{\partial z}\right) \cdot\left(x^3, y^3, z^3\right) \\
& 3 x^2+3 y^2+3 z^2 \\
\end{gather*}

I plug the divergence into the the Divergence formula, and convert the surface to spherical coordinates working out the bounds of each as I go in order to set up the triple integral.

\begin{gather*}
& \iint_S F \cdot ds \rightarrow \iiint_V 3 x^2+3 y^2+3 z^2 dV \\
& \quad \rho^2=x^2+y^2+z^2 \\
& \quad \rho^2=1 \\
& \rho=1 \\
& x=\rho \sin \phi \cos \theta \\
& y=\rho \sin \phi \sin \theta \\
& z=\rho \cos \phi \\
& 0 \leqslant \phi \leqslant \pi \\
& 0 \leqslant \theta \leqslant 2 \pi \\
& 0 \leqslant \rho \leqslant 1 \\
& \int_{\theta=0}^{\theta=2 \pi} \int_{\phi=0}^{\phi=\pi} \int_{\rho=0}^{\rho=1} 3\left(\rho^2 \sin ^2 \phi \cos ^2 \theta\right)+3\left(\rho^2 \sin ^2 \phi \sin ^2 \theta\right)+3\left(\rho^2 \cos ^2 \phi\right) \rho^2 \sin \phi d \rho d \phi d \theta \\
\end{gather*}

I then simplify the integrand as much as I can in order to simplify the integration. I use factoring and the trig identity cos^2x + sin^2x = 1 mulitple times to collapse the intergrand to $3\rho^4sin \phi$ and then proceed through a standard triple integral

\begin{gather*}
& \int_{\theta=0}^{\theta=2 \pi} \int_{\phi=0}^{\phi=\pi} \int_{\rho=0}^{p=1} 3 \rho^4 \sin \phi\left(\left(\sin ^2 \phi \cos ^2 \theta\right)+\left(\sin ^2 \phi \sin ^2 \theta\right)+\cos ^2 \phi\right) d \rho d \phi d \theta \\
& \int_{\theta=0}^{\theta=2 \pi} \int_{\phi=0}^{phi =\pi} 3 \rho^4 \sin \phi\left(\sin ^2 \phi\left(\cos ^2 \theta+\sin ^2 \theta\right)+\cos ^2 \phi\right) d \rho d \phi d \theta \\
& \int_{\theta=0}^{\theta=2 \pi} \int_{\phi=0}^{\phi=\pi} \int_{\rho=0}^{\rho=1} 3 \rho^4 \sin \phi\left(\sin ^2 \phi(1)+\cos ^2 \phi\right) d \rho d \phi d \theta \\
& \int_{\theta=0}^{\theta=2 \pi} \int_{\phi=0}^{\phi=\pi} \int_{\rho=0}^{\rho=1} 3 \rho^4 \sin \phi d \rho d \phi d \theta \\
& \int_{\theta=0}^{\theta=2 \pi} \int_{\phi=0}^{\phi=\pi} 3\left[\frac{1}{5} p^5 \sin \phi \right]_{\rho=0}^{\rho=1} d \phi d \theta \\
& \int_{\theta=0}^{\theta=2 \pi} \int_{\phi=0}^{\phi=\pi} \frac{3}{5} \sin \phi d \phi d \theta \\
& \int_{\theta=0}^{\theta=2 \pi} -\frac{3}{5}[\cos \phi]_{\phi=0}^{\phi=\pi} d \theta \\
& \int_{\theta=0}^{\theta=2 \pi} -\frac{3}{5}[-1-1] d \theta \\
& \int_{\theta=0}^{\theta=2 \pi} -\frac{3}{5}(-2) d \theta \\
& \int_{\theta=0}^{\theta=2 \pi} \frac{6}{5} d \theta \\
& {\left[\frac{6}{5} \theta\right]_{\theta=0}^{\theta=2 \pi}=\frac{12 \pi}{5}}
\end{gather*}


## Q4 — Conceptual

In one or two sentences each:

(a) What is the physical interpretation of the divergence of a vector field?

(b) What is the physical interpretation of the curl?

(c) Green's Theorem, Stokes' Theorem, and the Divergence Theorem are all instances of the same general principle. State that principle informally.

**Your answer:**

(a) divergence is the measure of net exit or entry to very small region around a point. Positive indicates there is more leaving the region (i.e. a source) and negative indicates more entering a regiona (i.e. a sink) if the result is zero then it is considered incompressible. In my mind I think of flow into a sink hole and flow out of a hole ( a fountain like source)

(b) curl represents the tendency to rotate a "paddle" whose axis is  normal to the region on the surface a clockwise tendency to rotate is positive and a counter clockwise is negative

(c) Greens/Stokes/Divergence leverage a realtionship between the curl and divergence of a region bounded by a closed curve in that the calculation of the curl and divergence of the interior region is the same as the boundary. This is because the interior "regions" edges cancel out leaving only the boundary edges as contributing the result. This is usefull as it allows us to manipulate an integral in order to make it easier (or possible) if one perspective isnt practical (and we can go in both directions.
